Ya por ultimo, recordemos que las preguntas que se realizaron en la encuentas eran calificadas en $ 1-5 $ y algunas de las respuesta tiene valores fuera de dicho rango y por ello debemos de cambiar dichos valores; pero como ???

Con la siguiente regla:
- Si el valor es mayor que $5$, suponemos que los respuesta a esa pregunta es es totalmente satisfactoria y por ende sera marcaremos en $5$.

- Todo lo contrario, si el valor es menor que cero

- Los valores Nulls los dejamos asi como esta *(los que son cero esta excento a esta regla)*.

Las columnas que infringen el rango de valores para responder son las siguiente:
**E10**, **N10**, **C8**, **03** Y **O9** *(por el analisis de datos realizado en la practica anterior)*

```python
# Almacenamos los nombre de las columnas
columnas_corregir = ['E10', 'N10', 'C8', 'O3', 'O9']

# Checa cada uno de los registro  y plica la regla que hemos dicho
for col in columnas_corregir:
    tabla_new[col] = np.where(tabla_new[col] > 5, 5, tabla_new[col])  # valores > 5 → 5
    tabla_new[col] = np.where(tabla_new[col] < 0, 1, tabla_new[col])  # valores < 0 → 1
```

En la teoria nos dice que con cubrir el mas de $70\%$ de **la varianza explicativa del PCA** es mas que suficiente para determinara el numero de componenetes ques que necesitamos. Dicha varianza sera representada por lo por los valores propios que si lo graficamos en **una grafica de Codo**

```python
# Establecemos el tamano del lienzo
plt.figure(figsize = (16, 5))

# los valores a tomar de la grafica: x,y repectivamente
x = range(1, len(pca.explained_variance_ratio_) + 1)
y = pca.explained_variance_ratio_             # Valores propios

# sera una grafica de linea que una los puntos "y"
plt.plot(x, y, marker='o', linestyle = '--', color ='b')

# configuramos los elementos de la grafica
plt.title('Gráfica de Codo - Varianza Explicada')
plt.xlabel('Número de Componentes Principales')
plt.ylabel('Proporción de Varianza Explicada')

# Mostramos la grafica
plt.grid(True)
plt.show()
```

Se le llama **Grafica de codo** ya que el punto que "quiebra" y forma el codo no indicaria la cantida de componentes que necesitamos que expresa el $88 \%$ de la varianza....

Para este caso diremos que con **25** componentes son mas que suficientes; ya que si vemos la propocion que ofrece cada uno de esto componentes....

```python
# ================== Sumamos Toda la Varianza =========================
# aqui estara la suma y empieza con 0
Sumvp1 = 0

# el "for" hara la suma totalmente
for i in range(0,len(y)):
  Sumvp1 += y[i]


# ================== Vemos la proporcion de varianza que ofrece =========================

# con esto podemos calcular las proporciones de cada componente.
contador = 0

# este variable suma la varianza hasta el caso presente
PT = 0

# el bucle " no ayuda a calcular las proporcion"
while contador <= 24:
  P1 = (y[contador] / Sumvp1) * 100
  print( f"La Proporcion del PC{contador + 1} es: {P1: .3f}%" )
  contador += 1
  PT += P1

print( f"Representa una variabilidad del {PT: .3f}%" )
```

Existe algo que se llama **Dendrograma** sera una grafica que estar uniendo los datos con una linea deacuerdo a la cercania que tiene los datos por jerarquias y se agrupara por **ramas**.

El numero de remas finales sera el numero de Clusterings que debemos formar para este modelo.

Para este caso vamos **estandarizar** que los procesos sean mas ligeros...

```python
# importamos la funcion de estandarizacion
from sklearn.preprocessing import StandardScaler

# Escalar los datos
scaler = StandardScaler()
tabla_escalada = scaler.fit_transform(tabla_new_4)
```

Pasando a la grafica:

```python
#==================== Preparamos los datos para la grafica ===================

# Importamos 2 funscion: dedrogram que sera la grafica y la funcion linkage que nos ayuda en la jerarquias
from scipy.cluster.hierarchy import dendrogram, linkage


# Tambien importamos la distancias
from scipy.spatial.distance import pdist

# Calcular las distancias usando la métrica de Manhattan
distancias = pdist(tabla_escalada, metric = 'cityblock')

# Calcular el linkage usando como parametro el promedio
Z = linkage(distancias, method = 'average')
```

```python
#==================== Grafica del Dedrograma ===================

# Establecemos el tamano del Lienzo
plt.figure(figsize = (25, 8))

# Indicamos que datos a graficar
dendrogram(Z)

# Configuramos lo elementos de la grafica
plt.title("Dendrograma con distancia Manhattan")
plt.xlabel("Observaciones")
plt.ylabel("Distancia (Manhattan)")

# Visualizamos la tabla
plt.show()
```
