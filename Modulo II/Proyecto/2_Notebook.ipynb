## **Que es una V.A.**

Sea un **Espacio de Probabilidad** $( \Omega, \mathcal{F}, \mathbb{P} )$. Una V.A. es una funcion que describe ha una evento denotado $X$ donde:

$$ X:\Omega \rightarrow \mathbb{R} $$

Asignandole a una **"identitificacion numerica"** a cada uno de los resultados de $\Omega$ *(como se observa en la imagen)*.

image.png

Como sabemos que algunas **V.A.** no abarcar a todo $\mathbb(R)$ y toma a algunos valores numericos; por lo que tendremos 3 tipo **V.A.**:

*   *CONTINUAS:* Sera las **V.A.s** que contemple un rango $\mathbb{R}(X)$ sobre todo los reales o con intervalos *(cerrados, abiertos, etc.)* $(a,b) \in \mathbb{R}$.

*   *DISCRETAS:* Sera las **V.A.s** que utilizaran un conjunto de datos finitos del tipo numerico *(puede ser $\mathbb{N}$ o $\mathbb{Z}$)*.

*   *MIXTAS:* Es un V.A. que por ciertas parte de ella sera Continua y en otras Discreta.

## **Funciones de Probabilidad**

Una **funcion de probabilidad** sera la manera en la que podemos calcular de forma eficaz la probabilidad de alguno de los resultados de la **V.A.** La contruccion de la esta funcion estra respaldada por un analisis cautelos de como podemos buscar los resultados de $X$ y colocarlos en la funcion de forma numerica.

Esta puede ser evaluadas de 2 forma posibles...

### **En un valor numerico de $\mathbb{R}(X)$**


 Las denotaremos como:

$$ \mathbb{P}(X = x) = f_X(x) $$

Que dependiendo del tipo de **V.A.** recibiran un nombre:

*   $\color{Green}{\text{Funcion de Masa de Probabilidad}}$ *(Si es del tipo **Discreto**)* y se abreviaran con **f.M.P**

*   $\color{Green}{\text{Funcion de Densidad de Probabilidad}}$ *(Si es del tipo **Continuo**)* y se abreviaran con **f.D.P**


### **En un rango de numeros e menores o mayores que un numero de $\mathbb{R}(X)$ y las denotaremos como:**

Las denotaremos como:

$$ \mathbb{P}(X \leq x) = F_X(x) $$

$$ \mathbb{P}(X > x) = 1 - \mathbb{P}(X \leq x) = 1 - F_X(x) $$

$$ \mathbb{P}(x_1 \leq X \leq x_2) = F_X(x_2) - F_X(x_1) $$

Que sin importar el tipo de **V.A.** lo nombraremos como: $\color{Green}{\text{Funcion de Probabilidad Acumulada}}$ y se abreviaran con **F.P.A**

Aunque representacion cambia dado a la **V.A.**, esto nos quedaria de la forma:

*  Caso *DISCRETO:* $ F_X(x) = \sum^x_{min{\mathbb{R}(X)}} f_X(x) $

*  Caso *CONTINUO:* $ F_X(x) = \int^x_{- \infty} f_X(x) $


Algunas propiedades que tiene $F_X(x)$ son las siguientes:

*   $ \lim_{x \rightarrow \infty} F_X(x) = 1 $

*   $ \lim_{x \rightarrow -\infty} F_X(x) = 0 $

*   $ F_X(x) $ no es **Decreciente**

*   $ F_X(x) $ es **Monotona**

**Observaciones:** Hay que tener en cuenta que las desiguladades nos ayuda a ver el intervalo que deseamos calcular. En el caso de **F.D.P** no tendra problema alguno dado a que $\mathbb{R}(X)$ es de forma *continua* ; el detalle esta en las **F.M.P** dado a la *descontinuidad* por lo que hay tomarlo por pinzas.

## **Valor Esperanza de la V.A.**

Tambien como **Esperanza Matematica** consiste en una promedio ponderado entre los valores $x_i \in \mathbb{R}(X)$ y $\mathbb{P}(X = x_i)$ los cuales no estaria indicando 2 cosas:

*   El Valor Medio de toda la **V.A.**
*   El numero mas esperado de toda la **V.A.** ya que hay $50 \%$ de ocurrir.

Por que es claro que varia deacuerdo a que tipo de **V.A.** que deseamos calcular...

*   Si es *DISCRETA:*
$$ \sum_{x \in \mathbb{R}(X)} x \cdot f_X(x) $$

*   Si es *CONTINUA:*
$$ \int_{- \infty}^{\infty} x \cdot f_X(x) dx $$

**Notacion:** $\mathbb{E}[X]$

## **Valor de Dispersion de la V.A.**

Conocido como **Varianza** consiste otro *Esperanza* , pero con la idea es ver que tanta dispersion estan los datos de media $\mu = \mathbb{E}[X]$...

Por definicion lo podemos representar de 2 forma:

*   $\mathbb{E}[(X - \mu )^2]$

*   $\mathbb{E}[X^2] - \mathbb{E}[X]^2$

**Notacion:** $Var(X)$

## **Funcion Generadora de Momentos**

Los **Momentos  de la V.A. $X$** consisten en una serie de *valores esperados* que nos ayuda a dar mas profundidad y descripcion de la funcion de probabilidad que este ocupando. Si bien ya hemos visto de los primeros 2 momentos:

*   $\mathbb{E}[X^1]$ *(La esperanza Matematica)*
*   $\mathbb{E}[X^2]$ *(Termino que ayuda a construir la Varianza)*

Y hay mas de Informacion con los momentos...

*   $\mathbb{E}[X^3]$ *(asimetría **skewness** de $f_X(x)$)*
*   $\mathbb{E}[X^4]$ *(**Curtosis** de $f_X(x)$)*

Y para continuar con el estudio de la estadistica si necesitamos los **k-esimos
 Momentos de X**...

Por $\color{blue}{\text{El Teorema del Estadistico Inconciente}}$ podemos calcular la esperanza matematica para cualquier funcion $g(\cdot)$ continua, por que por definicion tenemos que **Los Momentos de $X$** son:

*   Caso *DISCRETO:*

$$ \mathbb{E}[X^k] = \sum_{x \in \mathbb{R}(X)} x^k \cdot f_X(x) $$

*   Caso *CONTINUA:*

$$ \mathbb{E}[X^k] = \int_{- \infty}^{\infty} x^k \cdot f_X(x) dx $$

Pero el realizar este calculo se puede volver complicado conforme $k$ se vuelve mas grande, por que vamos ha crear un funcion que nos ayude a "*generar*" los momentos de $X$ rapidamente..

Sea $t \in \mathbb{R}$, definimos a $M_X(t) = \mathbb{E}[e^{tx}]$ como **La Funcion Generadora de Momentos de la V.A. $X$** con el cual nos devuelve **Los Momentos de $X$** ; siempre y cuando resolvamos la siguiente igualdad:

$$ \mathbb{E}[X^k] = \frac{d^k}{dx^k} \left( M_X(t) \right) \mid_{t = 0} $$

## **Funcion Caracteristica**

Esta funcion se trata en si **La Transformada de Fouir** para $X$ de la cual es un termino que existe para cualquier **V.A.** y tiene uso mas teoricos que practicos. Ya que tiene el objetivo de contruir a $f_X(x)$ por medio de su $\mathbb{R}(X)$.

Por definicion tenemos que **La Funcion Caracteristica** es:

$$ \varphi_X(w) = \mathbb{E}[e^{iwx}] $$

**Observaciones:**

*   El hecho de que $\varphi_X(w)$ exista que todo gira en la ***Formula de eular*** una expresion casi omnipresente de cada area de la matematicas...

$$ e^{iwx} = \cos(wx) x + i \operatorname{sen}(wx) $$

*   Podemos calcular **Los Momentos de V.A. $X$** de la siguiente forma:

$$ \mathbb{E}[X^k] = \frac{1}{i^k} \cdot \frac{d^k}{dx^k} \left( \varphi_X(w) \right) \mid_{t = 0} $$

*   $\varphi_X(w)$ se utiliza para la demostracion del $\color{blue}{\text{Teorema de Limite Central}}$ *(**TLC**)*

## $\color{purple}{\text{Variable aleatoria Uniforme discreta}}$

Una variable aleatoria $X$ se dice que tiene **Distribución Uniforme Discreta** si su **f.m.p** esta dada por:

$$ f_{X}(x) = \frac{1}{n} ~ \mathbb{I}_{\{1,2,...,n\}}(x) $$

*		**Notación.** $X \sim \text{Unif}(x_1,x_2,...,x_n)$

La **Función de Distribución Acumulativa** *(**F.D.A**)* se define como:

$$ F_{X}(x) = \frac{ \lfloor x \rfloor }{n} ~ \mathbb{I}_{\{1,2,...,n\}}(x) $$

El **Valor Esperado** tiene la siguiente forma:

$$ \mathbb{E}[X] = \frac{n+1}{2} $$

La **Dispersion de $X$** tiene la siguiente forma:

$$ Var(X) = \frac{n^2 - 1}{12} $$

La Funcion Generadora de Momentos *(**F.G.M**)* esta determinada de la siguiente forma:

$$ M_X(t) = \frac{e^{t}(1 - e^{nt})}{n(1 - e^t)} $$

### **Ejemplo y Grafica**

Vamos a graficar un ejemplo de como es la distribucion Uniforme Discreta

```python
# Definir el rango de la distribución uniforme discreta [low, high)
low = 0 #límite inferior
high = 10 #límite superior

#Crear los valores posibles dentro del rango
x = np.arange(low, high)

#Calcular la función de masa de probabilidad
pmf = np.full_like(x, 1/ (high-low) , dtype = float )

#Para crear una figura más grande
plt.figure(figsize=(10, 5))

#Graficar pmf
plt.bar(x, pmf, width=0.2, color='b', edgecolor='black')
plt.xlabel('Valores')
plt.ylabel('Probabilidad')
plt.title('Función de densidad de una uniforme discreta')
plt.grid(True)
plt.show() #mostrar el gráfica
```

## $\color{purple}{\text{Variable aleatoria Bernoulli}}$

Una variable aleatoria $X$ se dice que tiene **Distribución Bernoulli** si su **f.m.p** esta dada por:

$$ f_{X}(x) = p^x(1-p)^{1-x} ~ \mathbb{I}_{\{0,1\}}(x) $$

*  **Notación:** $X \sim Ber(p)$

**Observacion:**
*   Notemos que el objetivo de este **f.m.p** es medir la probabilidad de un ensayo bajo las condiciones de $p$..

La **Función de Distribución Acumulativa** *(**F.D.A**)* se realiza por definicion...


El **Valor Esperado** tiene la siguiente forma:

$$ \mathbb{E}[X] = \sum_{x = 0}^{1} x f_{X}(x) = 0\cdot (1-p) + 1 \cdot p = p $$

La **Dispersion de $X$** tiene la siguiente forma:

En base al ***Segundo Momento de la $X$*** que se define de la siguiente forma:

$$ \mathbb{E}[X^2] = \sum_{x = 0}^{1} x^2 f_{X}(x) = 0^2 \cdot (1-p) + 1^2 \cdot p = p $$

Por ende tendriamos que la varianza sera la siguiente:

$$ \text{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2  = p - p^2 = p(1-p)$$

La Funcion Generadora de Momentos *(**F.G.M**)* esta determinada de la siguiente forma:

$$ M_X(t) = 1 + p(e^t - 1) $$

### **Sobre el Parametro p**

Este modelo se aplica a un experiemento cuyo espacio muestral tiene dos resultados posibles:

$$ \Omega = \{\text{exito} , \text{fracaso}\} $$

y definimos por practicidad lo denotaremos como...

- $ X(\{\text{exito}\}) = 1 $          
- $ X(\{\text{fracaso}\}) = 0 $

Las probabilidades asociadas a este modelo...

- $ \mathbb{P}(\{\text{exito}\}) = p $
- $ \mathbb{P}(\{\text{fracaso}\}) = 1-p $

Donde $ 0 < p < 1 $ y este parametro la conoceremos como **"la probabilidad de exito".**

La función de masa de probabilidad *(f.m.p)*, está definida de la siguiente manera:

$$ f_X(x) = \mathbb{P}(X=x) = \begin{cases}
                                p^x(1-p)^{1-x}, \quad \text{si } x\in \{0,1\} \\\\
                                0, \quad \text{e.o.c}
                              \end{cases} $$

*  **Notación:** $X \sim Ber(p)$

Mediante la combinación de V.A. Bernoulli, es posible construir otras V.A. *(pero lop veremos mas adelante)*

La esperanza y varianza de $X$, esá dada por:

$$ \mathbb{E}[X] = \sum_{x = 0}^{1} x f_{X}(x) = 0\cdot (1-p) + 1 \cdot p = p $$

Para el caso de la varianza de la Bernoulli lo construiremos con en base al ***segundo momento de la $X$*** que se define de la siguiente forma:

$$ \mathbb{E}[X^2] = \sum_{x = 0}^{1} x^2 f_{X}(x) = 0^2 \cdot (1-p) + 1^2 \cdot p = p $$

Por ende tendriamos que la varianza sera la siguiente:

$$ \text{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2  = p - p^2 = p(1-p)$$

Vemos un ejemplo de como graficar la **FMP** de una bernoulli...

```python
from scipy.stats import bernoulli

#Definimos nuestra probabilidad de éxito
p=0.3

#Defino los posibles valores
x=[0,1]

#Calculamos la función de densidad
pmf = bernoulli.pmf(x, p)

print("La función de densidad = ", pmf)

#Graficamos la función de densidad
plt.bar(x,pmf,width=0.2)
plt.xlabel('Valores')
plt.ylabel('Probabilidad')
plt.title('Función de densidad de una bernoulli con p=0.3')
#plt.grid(True)
plt.show() #mostrar el gráfica

#Calculo de la esperanza
print("La esperanza es = ", bernoulli.mean(p))

#Calculo de la varianza
print("La varianza es = ", bernoulli.var(p))
```

## $\color{purple}{\text{Variable aleatoria Binomial}}$

Una variable aleatoria $X$ se dice que tiene **Distribución Binomial** si su **f.m.p** esta dada por:

$$ f_{X}(x) = \binom{n}{x}p^x (1-p)^{n-x} ~ \mathbb{I}_{\{0,1,...,n\}}(x) $$

*  **Notación:** $X \sim Bin(n,p)$

**Observacion:**

Los ensayos que se realizen, deben ser independientes entre ellas.

La **Función de Distribución Acumulativa** *(**F.D.A**)* se realiza por definicion...


El **Valor Esperado** tiene la siguiente forma:

$$ \mathbb{E}[X] = np $$

La **Dispersion de $X$** tiene la siguiente forma:

$$ \text{Var}(X) = np(1 - p)$$

La Funcion Generadora de Momentos *(**F.G.M**)* esta determinada de la siguiente forma:

$$ M_X(t) = \left( p e^t + (1 - p) \right)^n $$

**Ejemplo:**

Una moneda justa se tira 6 veces, donde la probabilidad de obtener sol es de $0.3$.

Sea $X$ el número de veces que cae sol, entonces en intuitivo que la V.A. $X$ se comporta como una Binomial con los parametros $ n = 6 $ y $ p = 0.3 $. Si quisieramos ver cual es la probabilidad de que salga tanto 2 o 3 soles en el experimento podemos calcularlo de la sig. forma:

- $ \mathbb{P}(X = 2) = \binom{6}{2}(0.3)^2 (1-p)^{6-2} = 0.3241 $
- $ \mathbb{P}(X = 3) = \binom{6}{3}(0.3)^3 (1-p)^{6-3} = 0.1852$

Tambien queremos ver cual es la probabilidad de que salgo entre 2 y 5 soles en los 6 lanzamientos...

$$ \begin{align*}
      \mathbb{P}(1 < X \leq 5) &= \mathbb{P}(X = 2) + \mathbb{P}(X = 3) + \mathbb{P}(X = 4) + \mathbb{P}(X = 5) \\
      &= 0.3241 + 0.1852 + \binom{6}{4} (0.3)^4 (0.7)^2 + \binom{6}{5} (0.3)^5 (0.7)^1 \\
      &= 0.5790
\end{align*}  $$

### **Sobre la distribucion**

La distribucion Bernoulli se trata de ver el resultado de un ensayo de algun experimento aleatorio; pero podemos generalizar este situacion con **"n-ensayos"** y ver la probabilidad que cierta cantidad de ensayo sean exitosos...

Decimos que una V.A. $X$ tiene distribución Binomial con parámetros:
*   $n$ *(que sera el numero de ensayos fijo)*
*   $p$ *(la probabilidad de exito de que un ensayo nos de un exito)*

Veamos como podemos ver la grafica de una Binomial....

```python
n = 25 # número de ensayos bernoulli
p = 0.8 # probabilidad de éxito
s= 100000 # número de muestras

#Vamos a generar numeros aleatorios que siguen una distribución binomial
binom_numeros = sps.binom.rvs(n,p,size=s)

#Creamos un histograma
plt.figure(figsize=(10,6))
plt.hist(
    binom_numeros,
    density=True, # Normaliza el area para que sea 1
    bins=len(np.unique(binom_numeros)), # número de barras del histograma
    color = "blue",
    edgecolor="grey"
)

plt.xlabel('Valores')
plt.ylabel('Probabilidad')
plt.title('Función de densidad de una binomial')
# plt.grid(True)
plt.show() #mostrar el gráfica
```

## $\color{purple}{\text{Variable aleatoria Poisson}}$



Una variable aleatoria $X$ se dice que tiene **Distribución Poisson** si su **f.m.p** esta dada por:

$$ f_{X}(x) = e^{-\lambda} \cdot \frac{\lambda^x}{x!} ~ \mathbb{I}_{\{0,1,2,...\}}(x) $$

*  **Notación:** $X \sim Poisson(\lambda)$

**Observacion:**

Es una distribución de probabilidad discreta que sirve para calcular la probabilidad de que ocurra un determinado número de eventos raros durante un selecto cantidad de datos a medir. Esta logica se puede trasladar a la un intervalo de datos *(puede ser tiempo, longitud, área, etc.)*, pero estos valores convertirian esta V.A. continua.

La **Función de Distribución Acumulativa** *(**F.D.A**)* se realiza por definicion...


El **Valor Esperado** tiene el mismo termino que la **Dispersion de $X$**; es decir....

$$ \mathbb{E}[X] = \lambda = \text{Var}(X) $$

La Funcion Generadora de Momentos *(**F.G.M**)* esta determinada de la siguiente forma:

$$ M_X(t) = e^{- \lambda (1 - e^\lambda)} $$

### **Ejemplo y una simulacion**

**Ejemplo:**

Supongamos que el número de accidentes que ocurre en un punto en un día tiene distribución Poisson con parámetro $ \lambda = 2 $,

- ¿cuál es la probabilidad de que en un día ocurran más de dos accidentes? *(Usaremos la forma "complementaria de la probabilidad")*
  
  $$ \begin{align*}
    \mathbb{P}(X > 2) = 1 - \mathbb{P}(X \le 2)
    &= 1 - \left[ \mathbb{P}(X = 0) + \mathbb{P}(X = 1) + \mathbb{P}(X = 2) \right] \\
    &= 1 - \left[ e^{-2}\frac{2^0}{0!} + e^{-2}\frac{2^1}{1!} + e^{-2}\frac{2^2}{2!} \right] \\
    &= 1 - e^{-2}[1+2+2] = 1-5e^{-2} = 0.3233
  \end{align*} $$

- ¿cuál es la probabilidad de que ocurran más de dos accidentes sabiendo que por lo menos ocurre uno? *(Usaremos la definicion de la "probabilidad condicional")*

  $$ \begin{align*}
    \mathbb{P}(X > 2 \mid X \ge 1) = \frac{\mathbb{P}(\{ X > 2 \} \cap \{X \ge 1\} )}{\mathbb{P}\{X \ge 1)\}} = @ = \frac{\mathbb{P}(\{X>2\})}{\mathbb{P}\{X \ge 1)\}} \\
    = \frac{.3233}{1-\mathbb{P}(X<1)} = \frac{0.3233}{1-e^{-2}} \frac{0.3233}{.8646} = 0.3739
  \end{align*} $$

**@:** Como los eventos $ \mathbb{P}(X > 2) $ e $ \mathbb{P}(X > 1) $ son eventos que estan intersectados entre si, por ende nos queda el evento mas grande que seria $ \mathbb{P}(X > 2) $

Podemos codificar estas probabilidades para verificar que son estos resultados...

```python
from scipy.stats import poisson

lambda_value = 2
prob_mayorq_2= 1- poisson.cdf(2, lambda_value)

print(f"P(X > 2): {prob_mayorq_2}")

# P(X >= 1) = 1 - P(X < 1) = 1 - P(X = 0)
prob_X_mayor_igual_1 = 1 - poisson.pmf(0, lambda_value)

# P(X > 2 | X >= 1) = P(X > 2 y X >= 1) / P(X >= 1)
# Como X > 2 implica X >= 1, entonces P(X > 2 y X >= 1) = P(X > 2)
prob_X_mayor_2_dado_X_mayor_igual_1 = prob_mayorq_2 / prob_X_mayor_igual_1

print(f"P(X > 2 | X >= 1): {prob_X_mayor_2_dado_X_mayor_igual_1}")
```

**Aproximación de Poisson a la Binomial**

En la teoria, **la distribución de Poisson** es una forma límite de **la distribución binomial**; es decir, es una buena aproximación cuando $n$ es suficientemente grande y $p$ suficientemente pequeña.

$ \color{Blue}{\textbf{Teorema (Poisson).}} $

Sean $S_{n}\sim Bin(n,p_{n})$ bajo el regimen $$\lim_{n \to \infty} np_{n}=\lambda > 0$$

Consideremos la siguiente sucesión de números reales:

$$ a_{j}(n,p_n) = \begin{cases}
											\binom{n}{j}(p_n)^{j}(1 - p_{n})^{n - j} ~ \text{ para} ~ j \leq n \\
											0 \text{ si} ~ j \geq n + 1
									\end{cases} $$

 Entonces:

 $$ \lim_{n\to \infty}a_{j}(n,p_n)=a_{j}=e^{-\lambda}\frac{\lambda^{j}}{j!} \ \ \forall j\in \mathbb{N}. $$

El teorema anterior implica que la distribución de Poisson ofrece un modelo probabilístico adecuado para todos aquellos experimentos aleatorios en los que las repeticiones son independientes unas de otras y en los que sólo hay dos posibles resultados: éxito o fracaso, con probabilidad de 	éxito pequeña, y en los que el interés se centra en conocer el número de éxitos obtenidos al realizar el experimento un número suficientemente grande de veces.

Empíricamente se ha establecido, que la aproximación se puede aplicar con seguridad si $ n \ge 100 $, $ p \le 0.01$ y $ np \le 20 $.

 **Ejemplo:**

Supongamos que la probabilidad de que un producto producido por cierta máquina es defectuoso es de $ 0.1 $. ¿Cuál es la probabilidad de que un lote de 10 productos contenga a lo más un producto defectuoso?

Sea $X$ el número de productos defectuosos y por intuicion sabemos que $X$ tiene una distribución binomial con parámetros $ n = 10 $ y $ p = 0.1 $, entonces ek proiblema lo podemos solucionar por medio de la sig. probabilidad:

\begin{align*}
  \mathbb{P}(X \le 1) &= \mathbb{P}(X = 0) + \mathbb{P}(X = 1) \\
  &= \binom{10}{0}(0.1)^{0}(0.9)^{10-0}+\binom{10}{1}(0.1)^{1}(0.9)^{10-1} \\
  &= 0.7361
\end{align*}

Ahora bien por el teorema anterior; Usando la distribución Poisson para resolver esta situacion. Pero el detalle esta en el tipo de $ \lambda $ usaremos, pero recordemos este parametro es *"el valor esperado"* por ende podemos decir que la esperanza de la distribucion binomial sera nuestro $ \lambda $; es decir:

$$ \lambda = E[X] = np = 10*(0.1) = 1 $$

Con lo anterior podemos volver a resolver la probabilida anterior...

$$ \mathbb{P}(X \le 1) = \mathbb{P}(X = 0) + \mathbb{P}(X = 1) = \frac{e^{-1}1^{0}}{0!} + \frac{e^{-1}1^{1}}{1!} = e^{-1} + e^{-1} = 0.7358 $$

Comparando ambos resultados, podemos ver que tiene la misma probabilidad *(a pesar que ambas se resolvieron con distribuciones distintas).*

**Simulacion sobre esta aproximacion**

Haremos una Simulación de la aproximación de la Binomial a la Poisson por medio una grafica...

```python
# -------------------------------------------------------------------------------------------
# Simulacion de la Aproximacion de la Binomial a la Poisson
# -------------------------------------------------------------------------------------------

# definimos las sig. variables...
param = 3           # Parametro de la Poisson que queremos aproximar
n = 1000            # Este es el número de ensayos en la distribución binomial
N = 5000            # Este es el número de simulaciones que realizaremos.

# Genera una muestra de N valores aleatorios de una distribución binomial con parámetros:
X = npr.binomial(n,param/n,N)

# Calcular la frecuencia relativa de los valores simulados
counts = np.bincount(X) / float(N)

# Crear un array de valores posibles
x = np.arange(len(counts))

# Calcular la función de masa de probabilidad (FMP) de la distribución Poisson teórica
f_x = sps.poisson.pmf(x, param)

# Graficamos la aproximacion que hemos desarrollado
plt.close()
plt.bar(x - 0.5, counts, width=1., label="ley empírica")
p2 = plt.stem(x, f_x, "r", label="ley teórica")
plt.legend()
plt.show()
```

## $\color{purple}{\text{Variable aleatoria Geometrica}}$

Una variable aleatoria $X$ se dice que tiene **Distribución Geometrica** si su **f.m.p** esta dada por:

$ \color{Green}{\text{Primera version}} $

La función de masa de probabilidades con parámetro $ 0 < p < 1 $ *(Probabilidad de exito)* que contara el numero de ensayos exitosos antes del primer fracaso:

$$ f_{X}(x) = p(1-p)^{x-1} ~ \mathbb{I}_{\{1,2,...\}}(x) $$

$ \color{Green}{\text{Segunda version}} $

La función de masa de probabilidades con parámetro $ 0 < p < 1 $ *(Probabilidad de exito)* que contara el numero de ensayos fracasados antes del primer exito:

$$ f_{X}(x) = p^{x}(1-p) ~ \mathbb{I}_{\{0,1,...\}}(x) $$

*		Notación: $ X \sim Geo(p) $

**Observacion:**

Esta **V.A.** $X$ se difiere de las demas ya que se desea ver que probabilidad del numero de veces en que los ensayos bernoulli nos de un estado antes de la primera aparicion del otro estado; dichos estados son: **exito** o **fracaso**...

La **Función de Distribución Acumulativa** *(**F.D.A**)* se define como:

$ \color{Green}{\text{Primera version}} $

$$ F_{X}(x) = 1 - (1 - p)^k ~ \mathbb{I}_{\{1,2,...\}}(x) $$

$ \color{Green}{\text{Segunda version}} $

$$ F_{X}(x) = 1 - (1 - p)^{k+1} ~ \mathbb{I}_{\{0,1,...\}}(x) $$


El **Valor Esperado** tiene la siguiente forma:

$ \color{Green}{\text{Primera version}} $

$$ \mathbb{E}[X] = \frac{1}{p} $$

$ \color{Green}{\text{Segunda version}} $

$$ \mathbb{E}[X] = \frac{1 - p}{1} $$

La **Dispersion de $X$** tiene la siguiente forma:

$$ \text{Var}(X) = \frac{1 - p}{p^2} $$

La cual aplica para ambas versiones de la distribucion.

La Funcion Generadora de Momentos *(**F.G.M**)* esta determinada de la siguiente forma:

$ \color{Green}{\text{Primera version}} $

$$ M_X(t) = p e^t \left( \frac{1}{1 - (1 - p) e^t} \right) $$

$ \color{Green}{\text{Segunda version}} $

$$ M_X(t) = \frac{p}{1 - (1 - p) e^t} $$

### **Ejemplo y Grafica**

**Ejemplo:**

Algunas de las propiedades que nos ofrece lo que las funciones de probabilidad es el hecho de qu respectar lo siguiente:

$$ \mathbb{P}(\Omega) = 1 $$

Por lo que, como sabemos que esta distribucion si es una funcion de probabilidad???

Esto es sencillo, que se sabe que una de las propiedades que cumple es que la suma *(por ser discreta)* de toda $f_X(x)$ sobre todo $R_X$ nos debe de dar $1$

Si $ 0 \le f_{X}(x) \le 1 $ para toda $x \in R_X$.

$$ \begin{align*}
	    \sum_{x \in R_{X}}f_{X}(x) &= \sum_{x = 1}^{\infty}(1-p)^{x-1}p \\
			&= p \sum_{y = 0}^{\infty}(1 - p)^{y} \\
			&= p \left( \frac{1}{1 - (1 - p)}\right) = 1
  \end{align*}$$

Para poder visualizar lo que esta distribucion lo codificaremos de la siguiente forma...

## $\color{purple}{\text{Variable aleatoria Binomial Negativa}}$

Una variable aleatoria $X$ se dice que tiene **Distribución Binomial Negativa** si su **f.m.p** esta dada por:

$ \color{Green}{\text{Primera version}} $

La función de masa de probabilidades con parámetro $ 0 < p < 1 $ *(Probabilidad de exito)* que contara el numero de ensayos realizados antes del  $r-exito$:

$$ f_{X}(x) = \binom{x - 1}{r - 1}p^{r}(1 - p)^{x - r} ~ \mathbb{I}_{\{r,r+1,...\}}(x) $$

*   **Notación.** $X \sim BinNeg(r,p)$.

**Observacion:**

*   Existe otra version de la distribucion *(que es contraparte a que veremos)*.

*   Parecida a la distribucion hermana, solo que difiere en proposito; mientras fijamos el numero de ensayos con $n$, esta se fijara el numero de exitos o fracasos que deseamos con $r$.

La **Función de Distribución Acumulativa** *(**F.D.A**)* se realiza por definicion...


El **Valor Esperado** tiene la siguiente forma:

$$ \mathbb{E}[X] = \frac{r}{p} $$

La **Dispersion de $X$** tiene la siguiente forma:

$$ \text{Var}[X] = \frac{r(1-p)}{p^2}.$$

La Funcion Generadora de Momentos *(**F.G.M**)* esta determinada de la siguiente forma:

$$ M_X(t) = \left( \frac{1}{1 - (1 - p) e^t} \right)^r $$

$$ \mathbb{P}(X = 4) = f_{X}(4) = \binom{4 - 1}{1 - 1} p^1 (1 - p)^{4 - 1} = \frac{3!}{3! * 0!} (.7)^1 (.3)^{3} $$

## $\color{purple}{\text{Variable aleatoria HiperGeometrica}}$

Una variable aleatoria $X$ se dice que tiene **Distribución HiperGeometrica** si su **f.m.p** esta dada por:

$$ f_{X}(x) = \frac{\binom{m}{x}\binom{N-m}{n-x}}{\binom{N}{n}} ~ \mathbb{I}_{\{0,1,2,...\}}(x) $$

*	**Notación:** $X\sim \text{Hiper}(N,n,m)$.

**Observacion:**

*   La definicion que recibe cada parametro son:
    * $N = $ Numero total de elementos.
    * $n = $ Numero de objeto tomado de la muestra *(sin reemplazar e sin orden).
    * $m = $ Numero de objetos tipo A de la selecion de tamana $n$.

*   Si $i \leq n ~ \text{y} ~ X \sim \text{Hiper}(n,N,m)$. Cuando $p = \frac{m}{N} ~ \text{y} ~ m,N$ son muy grandes con respecto a $n$ de $x$:

$$ \mathbb{P}(X = x) \approx \binom{n}{x}p^{x}(1 - p)^{n - x}.$$

**Situaciones que podemos resolver:**

*   Controles de calidad para procesos en los no es posible devolver la muestra.

## $\color{purple}{\text{Variable aleatoria Uniforme continua}}$

Una variable aleatoria $X$ se dice que tiene **Distribución Uniforme Continua** en el intervalo $ (a,b) $ tal que $ a,b \in \mathbb{R} $ y su **f.d.p** esta dada por:

$$ f_{X}(x) = \frac{1}{b-a} ~ \mathbb{I}_{[a,b]}(x) $$

*		**Notación.** $X \sim \text{Unif}(a,b)$ o $X \sim \text{Unif}[a,b]$

**Observaciones:**

* 	Esta distribucin es simétrica en es $(a,b)$.
* 	A los intervalos de la misma longitud contenidos en $(a,b)$ se les asigna la misma probabilidad. Esto se representa gráficamente con la probabilidad de que $X$ se encuentre en el intervalo $(s,t)$.
*		Se utiliza para simular los juegos del azar comunmete.

La **Función de Distribución Acumulativa** *(**F.D.A**)* se define como:

$$ F_{X}(x) = \frac{x-a}{b-a} ~ \mathbb{I}_{[a,b]}(x) $$

El **Valor Esperado** tiene la siguiente forma:

$$ \mathbb{E}[X] = \frac{a+b}{2} $$

La **Dispersion de $X$** tiene la siguiente forma:

$$ Var(X) = \frac{(b-a)^{2}}{12} $$

La Funcion Generadora de Momentos *(**F.G.M**)* esta determinada de la siguiente forma:

$$ M_X(t) = \frac{e^{bt} - e^{at}}{t(b - a)} $$

### **Ejemplos y Graficas**

Vemos como es el comportamiento de la $X$ visualmente *(de forma tradicional)*...

```python
# Parámetros de la distribución uniforme
a = 0  # límite inferior
b = 1  # límite superior

# Definimos las funciones de densidad y distribución
# Función de densidad (pdf) de la distribución uniforme
def uniform_pdf(x, a, b):
    return np.where((x >= a) & (x <= b), 1 / (b - a), 0)

# Función de distribución (cdf) de la distribución uniforme
def uniform_cdf(x, a, b):
    return np.where(x < a, 0, np.where(x > b, 1, (x - a) / (b - a)))

# Generar valores de x
x = np.linspace(-0.5, 1.5, 1000)

# Calcular la función densidad
pdf_values = uniform_pdf(x, a, b)

# Calcular la función de distribución
cdf_values = uniform_cdf(x, a, b)

# Graficar la PDF
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.plot(x, pdf_values, label='PDF', color='blue')
plt.xlabel('x')
plt.ylabel('Densidad de probabilidad')
plt.title('Función de Densidad (PDF)')
plt.grid(True)

# Graficar la CDF
plt.subplot(1, 2, 2)
plt.plot(x, cdf_values, label='CDF', color='orange')
plt.xlabel('x')
plt.ylabel('Función de distribución acumulada')
plt.title('Función de Distribución Acumulada (CDF)')
plt.grid(True)

plt.tight_layout()
plt.show()
```

## $\color{purple}{\text{Variable aleatoria Exponencial}}$

Una variable aleatoria $X$ se dice que tiene **Distribución Exponencial**, si su **f.d.p** esta dada por:

$$ f_{X}(x) = \lambda e^{-\lambda x} ~ \mathbb{I}_{(0,\infty)}(x) $$

*		**Notación.** $ X \sim \exp( \lambda )$

**Observaciones:**
*   Modelan tiempos de vida o de duracion de cierto organismo y aparatos

**Situaciones que podemos resolver:**
*   El tiempo transcurrido en un call center hasta recibir la primer llamada del día.
*   El  tiempo entre terremotos de una determinada magnitud.
*   Supongamos una máquina que produce hilo de alambre, la cantidad de metros de alambre hasta encontrar una falla en el alambre se podría modelar como una exponencial.

La **Función de Distribución Acumulativa** *(**F.D.A**)* se define como:

$$ F_{X}(x) = 1 - e^{-\lambda x} ~ \mathbb{I}_{(0,\infty)}(x) $$

El **Valor Esperado** tiene la siguiente forma:

$$ \mathbb{E}[X] = \lambda $$

La **Dispersion de $X$** tiene la siguiente forma:

$$ Var(X) = \lambda^2 $$

La Funcion Generadora de Momentos *(**F.G.M**)* esta determinada de la siguiente forma:

$$ M_X(t) = \frac{1}{1 - t\lambda} $$

### **Propiedad de pérdida de memoria**

Sea $X$ una V.A. tal que $X \sim exp(\lambda)$, entonces para cualesquiera $t, s \geq 0$ sucede lo siguiente...

$$ \mathbb{P}(X > t + s \mid X > t) = \mathbb{P}(X > s) = e^{-\lambda s} $$

$\color{red}{Demostracion:}$

Por la definicion de la probabilidad condicional:

$$ \mathbb{P}(X > t + s \mid X > t) = \frac{\mathbb{P}(X > t + s \cap X > t)}{\mathbb{P}(X > t)} = \frac{\mathbb{P}(X > t + s)}{\mathbb{P}(X > t)} $$

Es aqui donde utilizamos la **F.D.A**; pero recordando que si:

$$ \mathbb{P}(X > x) = 1 - F_X(x) = 1 - (1 - e^{- \lambda x}) = e^{- \lambda x} $$

Entonces...

$$ \frac{\mathbb{P}(X > t + s)}{\mathbb{P}(X > t)} = \frac{e^{- \lambda (t + s)}}{e^{- \lambda t}} = e^{- \lambda t - \lambda s + \lambda t} = e^{- \lambda s} $$

y esto por definicion queda: $ \mathbb{P}(X > s) $

### **Ejemplo**

**Ejemplo de puede resolver problematicas que los calculos no pueden hacer:**

Hipótesis natural para modelar las duraciones de vida de átomos radioactivas *(Rutherford y Soddy).*

Cada átomo radioactivo posee una duración de vida que sigue una ley exponencial. En este campo, el parámetro $\lambda$ se llama la constante de desintegración.

Si $ t \mapsto \rho(t) = \mathbb{P}(X > t), $ verifica que

$$ \rho(t+s)=\rho(t)\rho(s) $$

De manera que (derivando en $s$, con $s=0$)

$$ \rho^{\prime}(t) = -\rho(t)\lambda \qquad \ \lambda = -\rho^{\prime}(0)\geq 0. $$

Así...

$$ \rho(t) = e^{-\lambda t} \qquad \text{y} \qquad f(t) = \lambda e^{-\lambda t} $$

$$ \rho(0)=1 $$

**Ejemplo:**

entonces:

$$ 1 - \mathbb{P}(X \le 2) = 1 - F_{X}(2) = 0.1$$

Por lo que si lo vemos mas detenidamente a $ F_X(x) $...

$$ F_{X}(2) = 1-e^{\frac{-2}{\lambda}} =0.9 ~ \Rightarrow ~ e^{\frac{-2}{\lambda}}=0.1 ~ \Rightarrow ~ \frac{-2}{\lambda}=\ln(0.1) \Rightarrow \lambda=0.87 $$

Por lo tanto $ X \sim \exp(0.87) $.

## $\color{purple}{\text{Variable aleatoria Gamma}}$

Una variable aleatoria $X$ se dice que tiene **Distribución Gamma**, si su **f.d.p** esta dada por:

$$ f_{X}(x) = \frac{\lambda^{\alpha}}{\Gamma(\alpha)} \cdot x^{\alpha-1} \cdot e^{-\lambda x} ~ \mathbb{I}_{(0,\infty)}(x) $$

*		**Notación.** $ X \sim \Gamma(\alpha,\lambda) $ o $ X \sim Gamma(\alpha,\lambda) $

**Situaciones que podemos resolver:**
*   Modelar tiempos de espera.
*   Cuando se tiene interes sobre la acumulacion de una cantidad en un periodo de tiempo.
*   Se emplea como distribucion conjunta a priori en estadistica bayesiana.

La **Función de Distribución Acumulativa** *(**F.D.A**)* se define como:

$$ F_{X}(x) = \frac{\gamma(\alpha,x\beta)}{\Gamma{\alpha}} ~ \mathbb{I}_{(0,\infty)}(x) $$

Donde $\gamma(a,b)$ sera la **Funcion Gamma Incompleta**.

El **Valor Esperado** tiene la siguiente forma:

$$ \mathbb{E}[X] = \frac{\alpha}{\lambda} $$

La **Dispersion de $X$** tiene la siguiente forma:

$$ Var(X) = \frac{\alpha}{\lambda^2} $$

La Funcion Generadora de Momentos *(**F.G.M**)* esta determinada de la siguiente forma:

$$ M_X(t) = \left( \frac{\lambda}{\lambda - \lambda t} \right)^\alpha $$

### **Propiedades de la funcion Gamma**

**La distribución Gamma** se obtiene al considerar el tiempo que transcurre entre cierto número de ocurrencias de eventos que ocurren aleatoriamente en el tiempo...

La función gamma $\Gamma:(0,\infty)\to \mathbb{R}$ está definida como:

$$\Gamma(\alpha) = \int_{0}^{\infty}t^{\alpha-1}e^{-t}dt.$$

Algunas propiedades que nos ofrece esta funcion son las siguentes:

- $\Gamma(\alpha) < \infty$ para cualquier $\alpha > 0$.
- $\Gamma(\alpha + 1)=\alpha \Gamma(\alpha)$.
- Si $n \geq 1$ $\Gamma(n) = (n - 1)!$
- $\Gamma \left( \frac{1}{2} \right) = \sqrt{\pi}$.




La distribución gamma es esencial en varios campos por su capacidad para modelar tiempos de espera y eventos con tasas constantes. Sus aplicaciones incluyen:

  * $\textbf{Teoría de colas y procesos estocásticos:}$ Modela el tiempo de espera hasta el k-ésimo evento.
  * $\textbf{Confiabilidad y análisis de supervivencia:}$ Utilizada para tiempos de fallo y eventos críticos en medicina.
  * $\textbf{Hidrología y meteorología:}$ Aplica en la modelación de precipitaciones acumuladas y tamaños de gotas.
  * $\textbf{Procesamiento de imágenes y señales:}$ En el ajuste de modelos a datos de intensidades.
  * $\textbf{Finanzas:}$ Para rendimientos de activos que no siguen distribuciones normales.
  * $\textbf{Biología y ecología:}$ Para tasas de crecimiento y tiempos entre eventos biológicos.
  * $\textbf{Física:}$ Describe tiempos de decaimiento y distribuciones de energía.

**Ejemplo:**

Si se utilizan $10$ de estos aparatos en forma consecutiva, uno de ellos después de que el anterior ya no funciona. ¿Cuál es la probabilidad de que alguno de los aparatos estará funcionando después de $10,000$ horas?

$$ \mathbb{P}(X > 10000) = \int_{10000}^{\infty}\frac{(0.001)^{10}}{9!}x^{9}e^{-0.001x}dx = 0.4579.$$

### **Caso Particulares**

La particularidad de esta distribucion es que tiene la caracteristica de poder crear otras distribuciones como casos particulares...

 - Si $\alpha = 1$, $\lambda > 0 \Rightarrow$ exponencial.

- Si $\lambda = \frac{1}{2}$, $\alpha = \frac{k}{2}$, $k \in \mathbb{Z}^{+} \Rightarrow$ ji-cuadrada.

- Si $\lambda > 1$ y $\alpha > 1 \Rightarrow$ Erlang $\Rightarrow$ aplicaciones.


Hay otro caso particular que es interesantede ver:

Si $X \sim N(0,1)$, entonces $X^{2}\sim \Gamma \left( \frac{1}{2}, \frac{1}{2} \right)$.

En efecto, Para $z>0$, se tiene:

$$ F_{X^{2}}(z) = \mathbb{P}(X^2 \leq z) = \mathbb{P}(-\sqrt{z} \leq X \leq \sqrt{z}) = F_{X}(\sqrt{z}) - F_{X}(-\sqrt{z}) $$

Por lo tanto...

$$ f_{X^2}(z) = \frac{d}{dz}(F_{X^{2}}(z)) = \frac{d}{dz} ( F_{X}(\sqrt{z}) - F_{X}(-\sqrt{z}) ) = \frac{1}{2\sqrt{z}} f_{X}(\sqrt{z}) + \frac{1}{2\sqrt{z}} f_{X}(-\sqrt{z}) $$

Dado que la distribucion Normal es una funcion par entonces tenemos que:

$$ \frac{1}{2\sqrt{z}} f_{X}(\sqrt{z}) + \frac{1}{2\sqrt{z}} f_{X}(\sqrt{z}) =
\frac{2}{2\sqrt{z}} f_{X}(\sqrt{z}) = \frac{1}{\sqrt{z}} f_{X}(\sqrt{z}) $$

Y de acuerdo a la distribucion Normal estandar, veremos el desarrollo nos queda como...

$$ \frac{1}{\sqrt{z}} \frac{1}{\sqrt{2}} \frac{1}{\sqrt{\pi}} e^{\frac{-(\sqrt{z})^2}{2}} = (z)^{-\frac{1}{2}} \cdot \frac{\sqrt{1}}{\sqrt{2}} \cdot \frac{1}{\Gamma \left( {\frac{1}{2}} \right)} \cdot e^{-\frac{1}{2} z} = \frac{\left( \frac{1}{2} \right)^{\frac{1}{2}}}{\Gamma \left( {\frac{1}{2}} \right)} \cdot (z)^{\frac{1}{2} - 1} \cdot e^{-\frac{1}{2} z} $$

Por lo tanto: $X^{2}\sim \Gamma \left( \frac{1}{2}, \frac{1}{2} \right)$

## $\color{purple}{\text{Variable aleatoria Beta}}$


Una variable aleatoria $X$ se dice que tiene **Distribución Beta**, si su **f.d.p** esta dada por:

$$ f_X(x) = \frac{x^{\alpha-1}(1-x)^{\beta-1}}{B(\alpha, \beta)} ~ \mathbb{I}_{(0,1)}(x) $$

*   **Notación** $X\sim \text{Beta}(\alpha, \beta)$.

**Situaciones que podemos resolver:**

*   La distribución beta se utiliza en una variedad de campos, incluyendo:

  - $\text{Finanzas:}$ para modelar la variabilidad en tasas de retorno de inversiones.
  - $\text{Mercadotecnia:}$ para analizar proporciones de respuesta de consumidores.
  - $\text{Ciencias de la salud:}$ en la evaluación de la efectividad de tratamientos médicos.
  - $\text{Ecología:}$ para estimar proporciones en estudios de biodiversidad.

*   En la **Teoria de orden**.

*   Medir el grado de impureza en materiales.

**Observacion:**

*   si $\alpha = \beta = 1 ~ \Rightarrow ~ X \sim Unif(0,1)$


La **Función de Distribución Acumulativa** *(**F.D.A**)* se define como:

$$ F_{X}(x) = \frac{\Gamma(\alpha + \beta)}{\Gamma (\alpha) \cdot \Gamma(\alpha)} \cdot \int^x_0 U^{\alpha - 1} (1 - U)^{\beta - 1} dU $$


El **Valor Esperado** tiene la siguiente forma:

$$ \mathbb{E}[X] = \frac{\alpha}{\alpha + \beta} $$

La **Dispersion de $X$** tiene la siguiente forma:

$$ Var(X) = \frac{\alpha \beta}{(\alpha + \beta)^2 (\alpha + \beta + 1)} $$

La Funcion Generadora de Momentos *(**F.G.M**)* esta determinada de la siguiente forma:

$$ M_X(t) = 1 + \sum^\infty_{k = 1} \frac{t^k}{k!} \cdot \mathbb{E}[X^k] $$

### **Propiedades de la funcion Beta**

La distribución beta es una familia de distribuciones de probabilidad continua definida en el intervalo $[0,1]$. Es particularmente útil para modelar variables que representan proporciones y porcentajes.

La funcion beta se define de la siguiente forma:

$$ B(\alpha,\beta) = \int_0^1 t^{\alpha-1}(1-t)^{\beta-1} dt = \frac{\Gamma(\alpha) \Gamma(\beta)}{\Gamma(\alpha + \beta)} $$

**Ejemplo:**

Supongamos que un análisis sugiere que una nueva inversión tiene una alta probabilidad de éxito. Usando una distribución beta con $\alpha = 5$ y $\beta = 1$, entonces:

$$ f_{X}(x) = 5x^4, \quad \mathbb{E}[X] = \frac{5}{6}, \quad \text{Var}(X) = \frac{5}{252} $$

## $\color{purple}{\text{Variable aleatoria Normal}}$

La **Distribución normal** es una de las más importantes y de mayor uso tanto en *la teoría de la probabilidad*, como en *la teoría estadística*. También llamada **Distribución gaussiana** *(en honor a Gauss)* a quien se considera el padre de ésta distribución.

La importancia de la distribución normal, radica en el famoso $\color{red}{\text{Teorema central del limite}}.$ Fue descubierta por De *Moivre* en 1733 como un límite de la distribución binomial.


\Una variable aleatoria $X$ se dice que tiene **Distribución Normal**, si su **f.d.p** esta dada por:

$$ f_X(x) = \frac{1}{\sqrt{2\pi\sigma^{2}}} \cdot e^{-\frac{(x-\mu)^{2}}{2\sigma^{2}}} ~ \mathbb{I}_{(- \infty, \infty)}(x) $$

Donde:
*   $ \mu: $ La media aritmerica de nuestro $X$ del experimento aleatorio.
*   $ \sigma^{2}: $ La dispersion de los datos del $X.$

**Notación.** $X \sim N(\mu,\sigma^{2})$



**Observacion:**

- La curva tan caracteristica que presenta las distribucion se le conoce como **la campana de Gauss-Bell** *(que veremos mas adelante)...*

- La varianza es usada como una medida para comparar la dispersión en dos o más conjuntos de observaciones.

- La desviacion estandar sera nuestra $ \sigma = \sqrt{\sigma^2} $

- Una desviación estándar pequeña indica que los valores de la variable aleatoria se encuentran cercanos a la media.

- Una desviación estándar grande indica que los valores de la variable aleatoria se dispersan mucho con respecto a la media.

**Situaciones que podemos resolver:**

*   Por otra parte tiene permite modelar numerosos fenómenos naturales, sociales y psicológicos, por ejemplo:

    * 	Estatura.
    * 	Efectos de un fármaco.
    * 	Consumo de cierto producto por un grupo de individuos.
    * 	Coeficiente intelectual.
    * 	Nivel de ruido en telecomunicaciones.
    * 	Errores cometidos al medir ciertas magnitudes.

*   Juega un papel de suma importancia en **la inferencia estadística.**

*   Se ocupa como **Supuesto Principal** en los modelos de Regresion.



La **Función de Distribución Acumulativa** *(**F.D.A**)* se realiza por definicion...


El **Valor Esperado** tiene la siguiente forma:

$$ \mathbb{E}[X] = \mu $$

La **Dispersion de $X$** tiene la siguiente forma:

$$ Var(X) = \sigma^2 $$

La Funcion Generadora de Momentos *(**F.G.M**)* esta determinada de la siguiente forma:

$$ M_X(t) = e^{\mu t + \frac{\sigma^2 - t^2}{2}} $$

### **La estandarizacion de la Normal**

Mas conocidad como **la distribucion estandar** es la forma simple de **la distribucion Normal**, ya que el objetivo de esta es el forzar a que la distribucion normal que manejemos en otro; cuyo parametros son:

$$ \mu = 0 ~ ~ y ~ ~ \sigma^2 = 1 $$

Ya que no se esta realizando una cambio de valores de $X$, sino que "posicionando e reescalando" los datos *(para un calculo de probabilidad mas eficiente por medio de "tablas de probabilidad para normal estandar" e mas comodidad para otros temas)*

Pero para esto, es necesario el definir esta distribucion...

Sea $ X \sim N(\mu, \sigma^2) $. Definimos a la V.A. $Z$ tal que:

$$ Z = \frac{X - \mu}{\sigma} $$

Diremos que la funcion de densida de probabilidad de $Z$ se comporta de la siguiente forma:

$$ f_{Z}(z) = \frac{1}{\sqrt{2\pi}}e^{-\frac{z^2}{2}} ~ \mathbb{I}_{(- \infty, \infty)}(x) $$

*   **Notación.** $X \sim N(0,1)$

**Observaciones:**
Al operar con la distribucion estandar, es necesario usarlo de la siguiente forma:

$$ f_Z(z) = \Phi(z)  $$

Todo esto lo podemos verifivar por medio de la siguiente proposicion...

$ \color{blue}{\text{Proposicion:}} $

Sea $X \sim N(\mu,\sigma^{2})$, entonces:

$$ Z = \frac{X-\mu}{\sigma} $$

tiene una distribución gaussiana con media $0$ y varianza $1$, es decir, $Z\sim N(0,1)$.

$ \color{red}{Demostración:} $

Primero vamos a calcular la función de distribución acumulada  de $Z$:

$$ F_{Z}(z) = \mathbb{P}(Z \le z) = \mathbb{P}\left( \frac{X -\mu}{\sigma} \le z \right) = \mathbb{P}(X \le \sigma z + \mu) = F_{X}( \sigma z + \mu ) $$

Entonces, la función de densidad de $Z$ esta dada por:

$$ f_{Z}(z) = \frac{dF_{Z}(z)}{dz} = \frac{dF_{X}(\sigma z + \mu)}{dz} = \sigma f_{X}(\mu + \sigma z) = \frac{e^{-z^{2}/2}}{\sqrt{2 \pi}} $$

Pero podemos notar algo curioso en este resultado...

$$ \frac{e^{-z^{2}/2}}{\sqrt{2\pi}} = \frac{e^{-(z-0)^{2}/2}}{\sqrt{2 \pi 1^2}} $$

Dandonos a entender que: $ Z \sim N(0,1) $

## $\color{purple}{\text{Variable aleatoria Cauchy}}$

Una variable aleatoria $X$ se dice que tiene **Distribución Cauchy**, si su **f.d.p** esta dada por:

$$ f_X(x) = \frac{1}{\pi} \left[ \frac{\gamma}{(x - x_0)^2 + \gamma^2} \right] ~ \mathbb{I}_{(- \infty,\infty)}(x) $$

*   **Notación** $X\sim \text{Cauchy}(x_0, \gamma)$.

**Situaciones que podemos resolver:**

*   Para la solucion de **Ecuaciones diferenciales** relacionadas con las fisica.

*   En la evaluacion **Espectroscopica**.

**Observacion:**

*   Esta distribucion coincide con la **Distribucion T-students con algunos grados de libertad**

*   Cuenta con una version estandarizada y se le conocera como **La Distribucion Cauchy Estandar**; donde: $x_0 = 0 ~ y ~ \gamma = 1$ y se define de la forma:

$$ f_X(x) = \frac{1}{\pi (1 + x^2)} ~ \mathbb{I}_{(- \infty,\infty)}(x) $$

*   Los parametros tiene la funcionalidad sobre la grafica:

*   $x_0:$ es un parametro de localizacion.
*   $\gamma:$ es un pametro de escala.



image.png

La **Función de Distribución Acumulativa** *(**F.D.A**)*, se define como:

$$ F_{X}(x) = \frac{1}{\pi} \arctan \left( \frac{x - x_0}{\gamma} \right) + \frac{1}{2} ~ \mathbb{I}_{(- \infty,\infty)}(x) $$

Tanto el **Valor esperado**, la **Dispersion de $X$** y la **Funcion Generadora de Momentos** no existen para esta distribucion dado a sus complicaciones y la rareza que nos da.

## $\color{purple}{\text{Variable aleatoria Weibull}}$

Una variable aleatoria $X$ se dice que tiene **Distribución Weibull**, si su **f.d.p** esta dada por:

$$ f_X(x) = \frac{\alpha}{\beta^\alpha} \cdot x^{\alpha - 1} \cdot e^{- \left( \frac{x}{\beta} \right)^\alpha} ~ \mathbb{I}_{(0,\infty)}(x) $$

*   **Notación** $X\sim \text{Wei}(\alpha, \beta)$.

**Situaciones que podemos resolver:**

*   Analisis de supervivencia
*   Modelos para los procesos estocasticos relacionado en la fabricacion y distribucion de produccion
*   Meteodologia

**Observacion:**
*   Notemos que hemos utilizado parametros del alfabeto griego en las distribuciones anteriores, pero por lo general tiene esta funciones:

    *   $\alpha:$ Es el parametro de la Forma de la distribucion.
    *   $\lambda ~ o ~ \beta:$ Es el parametro de la escala de la distribucion.

*   Si $\alpha = 1 ~ \Rightarrow ~ X \sim \text{exp}(\beta ~ o ~ \lambda)$

La **Función de Distribución Acumulativa** *(**F.D.A**)* se define como:

$$ F_{X}(x) = 1 - e^{- \left( \frac{x}{\beta} \right)^\alpha} ~ \mathbb{I}_{(0,\infty)}(x) $$

image.png

El **Valor Esperado** tiene la siguiente forma:

$$ \mathbb{E}[X] = \beta \cdot \Gamma(1 + \frac{1}{\alpha}) $$

La **Dispersion de $X$** tiene la siguiente forma:

$$ Var(X) = \beta^2 \cdot \left[ \Gamma \left(1 + \frac{2}{\alpha} \right) - \left(1 + \frac{2}{\alpha} \right)^2 \right] $$

La Funcion Generadora de Momentos *(**F.G.M**)* esta determinada de la siguiente forma:

$$ M_X(t) = \sum^\infty_{k = 1} \beta^k \cdot \frac{t^k}{k!} \left( \frac{k}{\alpha} + 1 \right) $$

La ley de los Grandes Numeros *(**LGN**)* es una proposicion que se define de la siguiente forma:

### **Problema de la Aguja de Bufon**

Otra ejemplo de la utilidad que nos ofrece el *LGN* sera con el experimento a continuacion; donde este revela la existencia de $\pi$ anos antes de que se encontrara en las operaciones trigonometricas...

Para mas informacion: [**El problema que tiene el bufon con su agujas...**](https://www.estadisticaparatodos.es/taller/buffon/buffon.html)

```python
# Parámetros del problema
L = 1.0  # Longitud de la aguja
d = 1.0  # Distancia entre las líneas paralelas
num_trials = 100000  # Número total de lanzamientos
step = 1000  # Número de lanzamientos por paso para la estimación de pi

# Listas para almacenar los valores de pi estimados y los lanzamientos
pi_estimates = []
trial_steps = []
```

```python
# -------------------------------------------------------------------

#     Simulacion: Las agujas del Bufon con 10,000 lanzamientos

# -------------------------------------------------------------------


# Contador para cuántas agujas cruzan una línea
crossed_lines = 0

# Simulación de lanzamientos de la aguja y estimación de pi en pasos
for trial in range(1, num_trials + 1):
    # Posición del centro de la aguja
    center_position = np.random.uniform(0, d / 2)

    # Ángulo aleatorio entre 0 y pi (ángulo en radianes)
    angle = np.random.uniform(0, np.pi)

    # Distancia desde el centro de la aguja hasta la línea más cercana
    distance_to_nearest_line = L / 2 * np.sin(angle)

    # Verificamos si la aguja cruza una línea
    if center_position <= distance_to_nearest_line:
        crossed_lines += 1

    # Estimación de pi en pasos
    if trial % step == 0 and crossed_lines > 0:
        pi_estimate = (2 * L * trial) / (crossed_lines * d)
        pi_estimates.append(pi_estimate)
        trial_steps.append(trial)

print(f"Estimación de pi: {pi_estimate}")
```

```python
# Plot de la estimación de pi a lo largo de los lanzamientos
plt.figure(figsize=(10, 6))
plt.plot(trial_steps, pi_estimates, label="Estimación de π", color="blue")
plt.axhline(y=np.pi, color="red", linestyle="--", label="Valor real de π")
plt.xlabel('Número de lanzamientos')
plt.ylabel('Estimación de π')
plt.title('Convergencia de la estimación de π con el problema de la aguja de Buffon')
plt.legend()
plt.grid(True)
plt.show()
```

##  $ \color{orange}{\text{Teorema del Limite Central}} $

Recordemos de forma muy resumida que **El Teorema del Limite Central:**

$ \color{Red}{\text{Teorema:}} $

Sea $\{X_1, X_2, \dots, X_n\}$ una muestra de alguna poblacion que se comporte como **V.A.I.I.D** *(Variable Aleatoria Independientes e Identicamente Distribuidad)* con media $\mu$ y varianza $\sigma^2$ finitas *(existe)*.

Sea el **Promedio Muestral:**

$$ \bar{X}_n = \frac{1}{n} \sum_{i=1}^n X_i $$

Entonces, la variable aleatoria estandarizada:

$$ Z_n = \frac{\bar{X}_n - \mu}{\sigma / \sqrt{n}} $$

Converge en **Distribución a una Normal Estándar** ; es decir $ Z_n \xrightarrow{d} N(0, 1) $

Es decir,

$$ \lim_{n \to \infty} P\left( \frac{\bar{X}_n - \mu}{\sigma / \sqrt{n}} \leq z \right) = \Phi(z) $$

Cuando $n \to \infty$ *(por la **LGN**)*

**Importancia:**

*   Lo que nos dice es el hecho de que sin importan como se distribuye una **M.A.** podemos sacar y calcular parametros poblacionales ya que tiene una **Distribución a una Normal Estándar**.

*   Es una apoyo grande a la creacion de los **Estimadores** *(Especialmente los intervalos de confianza)*.

## $\color{purple}{\text{Metodos de generacion de numeros aleatorios rectangulares}}$

**Las variables generadas deben cumplir con:**
- Los valores generados deben ser independientes y estar idénticamente distribuidos.
- La secuencia generada debe ser lo más larga posible y ser reproducibles.

**Generación pseudoaleatoria:**

La generación de los números aleatorios rectangulares debe realizarse a través de relaciones **matemáticas de recurrencia**. Por esta razón se consideran **pseudoaleatorios**, ya que el proceso para generarlos es **determinístico** *(donde el experimento ya esta sometido a muchas condiciones, del cual ya se sabe que hay un resultado unico)*...

Hay 2 métodos que son los más utilizados para la generación. Ambos se basan en la siguiente definición:

$ \color{red}{\text{Definicion:}} $

Dos enteros $a$ y $b$ son **congruentes módulo $m$** si su diferencia es un múltiplo entero de $m$ y se expresa como:

$$𝑎 \equiv 𝑏 ~ \text{𝑚o𝑑} ~ 𝑚$$

Como consecuencia:

- $(a-b)$ es divisible entre $m$.
- $a$ y $b$ dan el mismo residuo al ser divididos entre $m$.

### **Metodo Congruencial Multiplicativo**

Generar una secuencia de números pseudoaleatorios uniformes en el intervalo (0, 1) usando la siguiente fórmula recursiva:

$$ n_{i+1} = a n_{i} ~ \text{mod} ~ m $$

- $m:$ debe ser tan grande como sea posible, dependiendo de los bits por palabra que maneje la computadora, descontando el bit del signo ($b$). Por lo tanto: $m = 2^b$

- $a:$ debe satisfacer que $a \approx 2^{(b+1)/2}$ y que $a \equiv \pm 3 ~ \text{mod} ~ 8$. La segunda expresión equivale a $( a-(\pm3 ) )$ es múltiplo de $8$.

- $n_0:$ entero positivo impar menor a $m$

- El periodo será de longitud: $m/4$

```python
# --------------------------------------------------------------------------------

# Simulacion de numeros pseudoaleatorio por este metodo...

# --------------------------------------------------------------------------------

# definimos los parametros de la funcion:
      #     n0 : valor inicial o semilla
      #      a : multiplicador
      #      m : modulo (2^b)
      #      n : el número de numeros que quiero generar

def multiplicativo(n0,a,m,n):
    secuencia = []                # Lista que guarda los valores
    ni = n0                   # iniciar con una semilla

# Con este bucle hacemos que realice la formula, la agrega a la lista y prosigue con la recursitividad
    for _ in range(n):
        ni = (a * ni) % m         # calculo el sig numero utilizando la formula recurrente
        secuencia.append(ni/m)      # normaliza entre 0 y 1
    return secuencia
```

**Ejemplo:**

Generame 10 numeros con una semilla inicial del $12345$ por medio de **Metodo Congruencial Multiplicativo**...

```python
# establecemos el valor de los parametros para la funcion: "multiplicativo"

m = 2**31
a = 65539     # a ≡ +-3 mod 8
n0 = 12345
n = 10
```

### **Método Congruencial Mixto**

Este método genera números pseudoaleatorios con la fórmula:
$$n_{i+1} = (a n_i + c) ~ \text{mod} ~ m$$

**Obsrvacion:** Se le llama “***mixto***” porque incluye una constante adicional $c$ y se estaria resolviendo con la multiplicacion e la suma.


- $m = 2^b$
- $a \approx 2^{(b-1)/2}$, $a \equiv 1 ~ \text{mod} ~ 4$
- $c$, $n_0$: enteros positivos impares $< m$
- Periodo: $m$

```python
# --------------------------------------------------------------------------------

# Simulacion de numeros pseudoaleatorio por este metodo...

# --------------------------------------------------------------------------------

# definimos los parametros de la funcion:
      #     n0 : valor inicial o semilla
      #      a : multiplicador
      #      c : constante impar
      #      m : modulo (2^b)
      #      n : el número de numeros que quiero generar

def mixto(n0,a,c,m,n):
    secuencia = []        # Lista que guarda los valores
    ni = n0               # iniciar con una semilla

# Con este bucle hacemos que realice la formula, la agrega a la lista y prosigue con la recursitividad
    for _ in range(n):
        ni = (a * ni + c) % m       # calculo el sig numero utilizando la formula recurrente
        secuencia.append(ni/m)          # normaliza entre 0 y 1
    return secuencia
```

```python
# Pasamos ver que nos devuelve este modelo
m = 2**31
a = 1103515245 # a cong 1 mod----
c = 12345
n0 = 42
n= 10
```

```python
secuencia_mixto = mixto(n0,a,c,m,n)
print(secuencia_mixto)
```

```python
# Generamos ambas secuencias para la MCM_1 y la Uniforme (usando los parametros anteriores)

# Generamos ambas secuencias
secuencia_mixto = mixto(n0,a,c,m,n)
secuencia_numpy = np.random.uniform(0,1,n)

# figura con Plotly
fig = go.Figure()

# Añadir los histogramas para los numeros pseudoaleatorios
fig.add_trace(go.Histogram(
    x=secuencia_mixto,
    nbinsx = 40,
    opacity = 0.6,
    name = 'Método mixto',
    marker_color='red'
))

# Añadir los histogramas para los valores que nos genera la distribucion uniforme
fig.add_trace(go.Histogram(
    x=secuencia_numpy,
    nbinsx = 40,
    opacity = 0.6,
    name = 'Numpy random uniform',
    marker_color='blue'
))

# elementos de la grafica anterior...
fig.update_layout(
    barmode='overlay', #superpone los histogramas
    title = 'Compararación: Método mixto -vs- Numpy',
    xaxis_title='valor generado',
    yaxis_title='Frecuencia',
    legend_title='Método',
    bargap=0.05
)

fig.show()
```

Estos metodos soin buenos a la hora de generar numeros psedoaleatoria, pero no es perfecto; cada uno de ellos tiene su ventaje e desventajas...


| **Método**     | **Fórmula**                            | **Ventajas**                                          | **Riesgos / Limitaciones**                                |
|----------------|----------------------------------------|-------------------------------------------------------|------------------------------------------------------------|
| Multiplicativo | $ n_{i+1} = a n_i ~ \text{mod} ~ m $           | Rápido, menos memoria                                 | Menor aleatoriedad, requiere \( a, m, n_0 \) bien elegidos |
| Mixto          | $ n_{i+1} = (a n_i + c) ~ \text{mod} ~ m $     | Mejor distribución, periodo completo posible          | Más complejo, pero más robusto                             |
| NumPy (`uniform`) | Motor moderno y validado           | Alta calidad, probado, rápido                         | Caja negra, sin control del generador interno              |

## $\color{purple}{\text{Metodo de la transformada inversa}}$

El método utiliza la función de distribución $F(x)$ de la distribución que se va a simular, que por definicion tenemos que:

$$F(x) = \int_{-\infty}^{x} f(t)\,dt$$

image.png

Como se sabe, los valores de $F(x)$ están en el intervalo $(0,1)$ al igual que los números rectangulares $U$.

El método genera un $U$ y trata de determinar el valor de la variable aleatoria para la cual $F(x)$ sea igual a $U$.

Si $U \in (0,1)$:
$$ F(x) = U \quad \Rightarrow \quad x = F^{-1}(U)$$

image.png

$\color{blue}{Teorema:}$

Sea $X$ una variable aleatoria real. Supongamos que su función de distribución $F$ es estrictamente creciente. Sea $U\sim \text{unif}[0,1]$. Entonces $F^{-1}(U)$ tiene la misma ley que $X$.

---

**Nota:** Por lo que $F$ es una biyección de $\mathbb{R}$ sobre $(0,1)$ y podemos denotar por $F^{-1}$ a su inversa.

---

Si $F$ no es estrictamente creciente, hemos visto que el teorema precedente sigue siendo válido bajo la condición de definir:

$$F^{-1}(u)=\inf\{ x\in \mathbb{R} : F(x)\geq u\}$$

La inversa generalizada de $F$.

### **Ejemplo:** $\text{Con la Distribucion Bernoulli}$

```python
# --------------------------------------------------------------------------------

# Simulacion de n valores de la bernoulli.

# --------------------------------------------------------------------------------

# Importamos la paqueteria "Random"
import random

# definimos una funcion y que nos devuelva 1 si se cumple la desigulada
def bernoulli(p):
    u = random.random()     # U -> Unif(0,1)
    return 1 if u <= p else 0
```

```python
# Simulamos n valores con p de éxito
p = 0.5

# Guardamos los resultados en uns lista
simulaciones = [bernoulli(p) for _ in range(100)]
print(simulaciones)
```

### **Ejemplo:** $\text{Con la Distribucion Exponencial}$

Sea $\lambda>0$ ; entonces a través del teorema, se puede generar una **V.A.** Exponencial.

Sea $U \sim Unif[0,1]$; entonces si $X \sim exp(\lambda)$, entonces:

$$ f_{X}(x) = \lambda e^{-\lambda x} ~ ~ ~ ~ ~ ~ ~ \text{y} ~ ~ ~ ~ ~ ~ ~ F_X(x) = 1-e^{-\lambda x} $$

Realizado la igualdad tenemos que:

$$ 1-e^{-\lambda x} = U ~ \Rightarrow ~ 1-U = e^{-\lambda x} ~ \Rightarrow ~ \ln(1-U) = -\lambda x ~ \Rightarrow ~ - \frac{\ln(1 - U)}{\lambda} + x $$

Por lo tanto tenemos que:

$$ -\frac{\ln(U)}{\lambda} \sim Exp(\lambda) $$

Pasemos a verlo graficamente los numeros psedoaleatorios que genera...

```python
# --------------------------------------------------------------------------------

# Simulacion de n valores de la Exponencial.

# --------------------------------------------------------------------------------


# Traemos la paqueteria para la simulacion
import numpy as np
import plotly.graph_objects as go

# Semilla
np.random.seed(123)

# Parámetro de la exponencial
lambd = 4

# La cantidad de numeros a conseguir
n = 10000

# Simulación usando transformada inversa
uniformes = np.random.random(n)
exponenciales = -np.log(uniformes) / lambd     # Usamos la formula que determinamos

# Curva teórica
x_vals = np.linspace(0, exponenciales.max(), 300)
y_vals = lambd * np.exp(-lambd * x_vals)

# Gráfico
fig = go.Figure()

fig.add_trace(go.Histogram(
    x=exponenciales,
    nbinsx=40,
    histnorm='probability density',
    marker_color='skyblue',
    name='Simulación'
))

fig.add_trace(go.Scatter(
    x=x_vals,
    y=y_vals,
    mode='lines',
    name='Densidad teórica',
    line=dict(color='darkblue')
))

fig.update_layout(
    title='Distribución Exponencial simulada vs teórica (λ = 1)',
    xaxis_title='x',
    yaxis_title='Densidad',
    bargap=0.05
)

fig.show()
```

### **Ejemplo:** $\text{Con la Distribucion Uniforme Continua}$

Sea $U \sim Unif[0,1]$; entonces si $X \sim Unif(a,b)$, entonces:

$$ f_{X}(x) =  \frac{1}{b-a} ~ ~ ~ ~ ~ ~ ~ \text{y} ~ ~ ~ ~ ~ ~ ~ F_X(x) = \frac{x-a}{b-a} $$

Realizado la igualdad tenemos que:

$$ \frac{x-a}{b-a} = U ~ \Rightarrow ~ x-a = U(b-a) ~ \Rightarrow ~ x = U(b-a) + a $$

Por lo tanto tenemos que:

$$ U(b-a) + a \sim Unif(a,b) $$

Pasemos a verlo graficamente los numeros psedoaleatorios que genera...

```python
# --------------------------------------------------------------------------------

# Simulacion de n valores de la Uniforme Continua.

# --------------------------------------------------------------------------------

# Traemos la paqueteria para la simulacion
import numpy as np
import plotly.graph_objects as go


# Parámetros de la uniforme
a, b = 2, 5
n = 10000

# Simulación por transformación lineal
uniformes = np.random.random(n)
uniforme_continua = a + (b - a) * uniformes

# Densidad teórica (constante)
x_vals = np.linspace(a, b, 300)
y_vals = np.ones_like(x_vals) * (1 / (b - a))

# Gráfico
fig = go.Figure()

fig.add_trace(go.Histogram(
    x=uniforme_continua,
    nbinsx=40,
    histnorm='probability density',
    marker_color='orange',
    name='Simulación'
))

fig.add_trace(go.Scatter(
    x=x_vals,
    y=y_vals,
    mode='lines',
    name='Densidad teórica',
    line=dict(color='red')
))

fig.update_layout(
    title='Distribución Uniforme Continua simulada vs teórica [2, 5]',
    xaxis_title='x',
    yaxis_title='Densidad',
    bargap=0.05
)

fig.show()
```

### **Ejemplo:** $\text{Con la Distribucion Cauchy Estandar}$

Sea $U \sim Unif[0,1]$; entonces si $X \sim Unif(a,b)$, entonces:

$$ f_{X}(x) = \frac{1}{\pi (1 + x^2)} ~ ~ ~ ~ ~ ~ ~ \text{y} ~ ~ ~ ~ ~ ~ ~ F_{X}(x) = \frac{1}{\pi} \arctan(x) + \frac{1}{2} $$

Realizado la igualdad tenemos que:

$$ \frac{1}{\pi} \arctan(x) + \frac{1}{2} = U ~ \Rightarrow ~ \frac{1}{\pi} \arctan(x) = U - \frac{1}{2} ~ \Rightarrow ~ $$

$$ \arctan(x) = \pi \left( U - \frac{1}{2} \right) ~ \Rightarrow ~ x = \tan \left( \pi \left( U - \frac{1}{2} \right) \right) $$

Por lo tanto tenemos que:

$$ \tan \left( \pi \left( U - \frac{1}{2} \right) \right) \sim Cauchy(0,1) $$

Pasemos a verlo graficamente los numeros psedoaleatorios que genera...

```python
# --------------------------------------------------------------------------------

# Simulacion de n valores de la Cauchy.

# --------------------------------------------------------------------------------

# Traemos la paqueteria para la simulacion
import numpy as np
import matplotlib.pyplot as plt

# Generar una variable aleatoria Cauchy
def generar_cauchy(n):
    U = np.random.uniform(0, 1, n)  # n valores uniformes
    X = np.tan(np.pi * (U - 0.5))  # Aplicar la fórmula
    return X

# Generar 1000 muestras de la distribución Cauchy
muestras_cauchy = generar_cauchy(1000)

# Filtrar valores extremos para centrarse en una parte más visible
muestras_centrales = muestras_cauchy[np.abs(muestras_cauchy) < 10]

# Crear un histograma de las muestras filtradas
plt.hist(muestras_centrales, bins=100, density=True, alpha=0.7, color='blue')

plt.title('Histograma Acercado de la Distribución Cauchy')
plt.xlabel('Valor')
plt.ylabel('Densidad')
plt.show()
```
