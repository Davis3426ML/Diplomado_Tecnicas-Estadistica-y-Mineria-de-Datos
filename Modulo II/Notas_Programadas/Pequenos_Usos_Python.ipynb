# Algunos casos

Python en un programa de pseudointerprete que actualmente es muy usado por la mayoria por practicidad y por supuesto su enfoque a **la orientacion por objetos**; donde un objeto es una entidad que hace ciertas acciones por medio de los atributos que se le asigne.

Los objetos tienen atributos (miembros):

* Datos: información útil para el objeto
* Métodos (funciones): operaciones sobre el propio objeto
Los atributos son accesibles con la siguiente notación: `objeto.atributo`

Existen dos modos en que un objeto es usado en una función:

```python
# Pensándolo como un argumento explícito
a = -3
abs(a)
```

```python
# Llamandolo a uno de sus métodos:
a = 'hola'
a.upper()     # atributo método
```

Digamos que deseamos un objeto que es en si un numero imaginario. Sea $z = 1+3i\in\mathbb{C}$, entonces

```python
a = 1 + 3j          # j denota a la parte imaginaria del complejo
im = a.imag             # atributo dato (parte imaginaria)
re = a.real
conj = a.conjugate()

print(im)           # imprimimos las partes que componen nuestro numero imaginario
print(re)
print("La parte real de", a, "es", re)
print(conj)
```

## Rango (Range)

Generador de índice (etiquetas en Python), esto se llama `range(n)`

```python
range(10)
```

```python
for i in range(10):
    print(i)
```

```python
lista_2 = [i for i in range(10)]
lista_2
```

```python
lista_3 = [i for i in range(1,10)]
lista_3
```
