# **Probabilidad Laplaciana (clásica)**

Para un experimento aleatorio, se tiene que $\Omega$ es el espacio muestral.

Sea $A$ un evento, entonces

$$ \mathbb{P}(A) = \frac{\#A}{\#\Omega} $$

**EJEMPLO:**

Sea el experimento de lanzar una moneda 3 veces, de manera independiente (es decir, el $i$-ésimo lanzamiento no me afecta al $i+1$ lanzamiento). El espacio muestral de este experimento es:

$$ \Omega = \{A,S\} \times \{A,S\} \times \{A,S\} = \{A,S\}^3 $$
y tenemos que $\# \Omega = 2^3$.

Sean los eventos:

* $A$: todos los resultados son águila
* $S$: todos los resultados son soles
* $E$: Por lo menos un resultado es sol

**Calcular las probabilidades de estos eventos anteriores:**

**Solución:**

Para cada uno de los casos vamos a verlo como si fuera un comulo de subconjunto, pero antes; vamos a desarrollar $\omega$ por medio de una paqueteria clasica...

```python
# Vamos a crear el espacio muestral
import itertools                # importamos toda la paqueteria
from itertools import product   # De esto usamos la funcion "product" para hacer todas la posibles resultados del experimento

Omega = set( product({"A","S"},repeat = 3 ))    # en esta variable guardamos los resultados
Omega
```

```python
len(Omega) # cardinalidad de Omega
```

## Caso de $A$

Vamos a descomponer mi conjunto $A$ de la siguiente manera:

$$A = A_1 \cap A_2 \cap A_3 $$

donde $A_i$:  caé águila en el $i$-ésimo resultado, veamos cada uno de de los subconjunto...

```python
# Vamos a crear el evento A_1
A_1 = { om for om in Omega if om[0] == "A"}     # la primera coordenada es águila
A_1
```

```python
# Vamos a crear el evento A_2
A_2 = { om for om in Omega if om[1] == "A"}     # la segunda coordenada es águila
A_2
```

```python
# Vamos a crear el evento A_3
A_3 = { om for om in Omega if om[2] == "A"}     # la tercera coordenada es águila
A_3
```

Y siguiendo con las interseccion de los $A_i$ anteriores...

```python
# Vamos a crear el evento A
A = A_1.intersection(A_2.intersection(A_3))
A
```

```python
len(A)    # vemos la cardinalidad de nuestro evento A
```

```python
# Vamos a calcular la probabilidad de A
P_A = len(A) / len(Omega)
P_A
```

Yo quiero expresar mi resultado en fracción, esto se hace de la siguiente manera:

```python
from fractions import Fraction    # importamos la funcion "Fraction"

def p(event,space):
    p = Fraction(len(event),len(space))     # definimos una funcion con los parametros: 'E' e 'Omega'
    return p

prob = p(A,Omega)     # Utilizamos la funcion "p" en nuestro evento A
prob
```

# **Probabilidad Condicional**

Definimos la probabilidad de que ocurra un evento $A$ sabiendo que ocurre el evento $B$ de la siguiente forma:


$$ \mathbb{P}(A \mid B) = \frac{\mathbb{P}(A \cap B)}{\mathbb{P}(B)} = \frac{\# A\cap B}{\#B} ~ ~ donde ~ ~ \mathbb{P}(B)>0. $$

```python
# para ello vamos a crea un funcion que recree la definicion....
def prob_cond(event_1,event_2):
    return len(event_1 & event_2) / len(event_2)
```

Quiero calcular la probabilidad condicional de los sig. eventos:
  
  * $P:$ El primer lanzamiento es águila
  * $Q:$ Dos primeros resultados son águila

```python
# El primer lanzamiento es águila
P = { om for om in Omega if om[0] == "A"}
P
```

```python
# los primeros 2 lanzamientos es águila
Q = { om for om in Omega if om.count("A") == 2}
Q
```

Entonces podemos ver cuales son sus respectivas probabilidades...

```python
print("La probabilidad de P es:", p(P,Omega))
print("La probabilidad de Q es:", p(Q,Omega))
```

Y veamos como queda la probabilidad de ambos eventos...

```python
prob_cond(P,Q)
```

# **Independencia de eventos**



Decimos que dos eventos $A$ y $B$ son independientes entre si, si:

$$ \mathbb{P}(A \cap B) = \mathbb{P}(A)\mathbb{P}(B)$$

O estos nos estaria dado a entender:

$$ \mathbb{P}(A | B ) = \frac{\mathbb{P}(A \cap B)}{\mathbb{P}(B)} = \frac{\mathbb{P}(A) ~ \mathbb{P}(B)}{\mathbb{P}(B)} = \mathbb{P}(A) $$

```python
# Vamos a definir una funcion encargada de verificar si 2 eventos totalmente independientes..
def indep(event_1, event_2, space):
    return p(event_1 & event_2, space) == p(event_1, space) * p(event_2, space)
```

Observemos que tanto los eventos $P$ e $Q$ no son totalmente independientes, sino son eventos dependientes.

```python
indep(P,Q,Omega)
```

```python
# Vemos que probabilidad tiene D
p(D,Omega)
```

Definimos el evento $S_n$ como la suma de los resultados con valor $n$ dado por los 2 dados lanzados en el experimento; es decir, lo veriamos como:

$$ S_n = \{ (i,j)\in\Omega \mid i+j = n \}$$

Este $S_n$ lo vamos a simular por medio de una funcion que construiremos a continuacion...

```python
def S(n):
    Sn = { (i,j) for i in range(1,7) for j in range(1,7) if i+j == n }
    return Sn
```

```python
# Vamos a ver todo los casos posibles del evento "S_9"
S_nueve = S(9)
S_nueve
```

$$ \mathbb{P}(S(9)) = 4/36 $$

```python
len(S_nueve) / cardOm
```

# **Probabilidad Frecuentista**

***La probabilidad frecuencial*** (también conocida como probabilidad frecuentista)b refiere a qué tan probable resulta un suceso si un experimento se repite muchas veces.

Tenemos la hipótesis de que el espacio muestral es un conjunto finito, y podemos definir la probabilidad asociada a un evento $A$:

$$ f_n(A) = \frac{n(A)}{n}, ~ ~ donde: $$

*   $n(A)$ denota al número de veces que ocurre $A$
*   $n$ es el número de veces en que se realiza el experimento.

```python
# Se tiene el lanzamiento de monedas injustas (volados)

import random as rm  # importamos la paqueteria random

def volado(p):      # Definimos esta funcion que nos devuelva aguila o sol con cierta probabilidad
    return 'sol' if rm.random() < p else 'aguila'
```

```python
# Vemos que resultado nos da este volado a una probabilidad de 0.4
volado(0.4)
```

```python
# Creamos una variable que guarde los resultados de "10 volados" con la funcion anterior
a = [volado(0.4) for i in range(10)]
print(a)
```

```python
# contamos "los soles" que sale de los 10 volado anteriores..
a.count("sol")
```

```python
# Tenemo otro ejemplo de 10 volados de una moneda injusta, pero con una codigo mas simple.

resultados = []       # lista vacia

for i in range(10):
    resultado = volado(0.4)
    resultados.append(resultado)

print(resultados)
```

Por como nos define la probabilidad, es necesario el crear mas simulaciones que tener una probabilidad mas precisa de una evento. Por lo qu necesitas un codificacion mas eficiente para guardar los resultados de los volados...

```python
# Vamos a simular nuestra probabilidad frecuentista

p = [] # Lista para almacenar "la proporcion que salga sol" de cada simulacion

for i in range(1000): # Generamos 1000 simulaciones
    N = 10            # Cada experimento tendra el #numero de lanzamientos (N)
    resultados = []   # Restearemos la lista en cada simulacion

    for i in range(N):        # en "resultados" guardamos lo que salga en la funcion "volado()"
        resultado = volado(0.63)
        resultados.append(resultado)

    n_soles = resultados.count("sol") # Cuenta en #soles en "resultados"
    p_soles = n_soles / N             # Calcula la proporción

    p.append(p_soles)     # Dicho resultado sera almacenado en la lista "p"
```

```python
# Probabilidad promedio

p_soles_prom = sum(p) / 1000

print("Probabiliad {:.3f}".format(p_soles_prom))
```
