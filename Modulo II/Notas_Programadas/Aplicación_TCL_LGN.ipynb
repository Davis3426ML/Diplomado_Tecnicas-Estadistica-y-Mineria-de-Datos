# Simulación de la LGN

La ley de los Grandes Numeros (LGN) es una proposicion que nos dice que :

Para $ X_1,..., X_n $ una muestra aleatoria *(m.a.)*. Definimos $ M_n = \frac{\sum x_i}{n} $ como otra V.A. pero que conforme $n$ se va haciendo mas grande, nuestra $ M_n $ tiende mas a su valor esperado la distribucion que comparte la **m.a.**; es decir:

$$ M_n \rightarrow \mathbb{E}[X_i] $$

Simularemos un experimento aleatorio clasico; el lanzamiento de un dado y ver cual es el resultado...

$$ \Omega = \{ 1,2,3,4,5,6 \}$$

Si calculamos su media teoria $\mu$ del experimento seria la siguiente forma...

$$ \mathbb{E}[X] = \sum_{i=1}^n x f_X(x) = 1 \cdot \frac{1}{6} + 1 \cdot \frac{1}{6} + 2 \cdot \frac{1}{6} + 3 \cdot \frac{1}{6} + 4 \cdot \frac{1}{6} + 5 \cdot \frac{1}{6} + 6 \cdot \frac{1}{6} $$

$$ \mu = \mathbb{E}[X] = \frac{21}{6} = 3.5 $$

Veremos una simulacion de un tamano de 100 *(esto representa el tamano de nuestra m.a.)*...

## Con 100 Simulaciones.

```python
np.random.seed(42)            # Guardamos nuestro pseudoaleatoridad que haremos en la simulacion (que sea replicable despues)
lanzamientos = np.random.randint(1,7,100)     # Guardamos nuestro resultados de lanzamiento del dado

print("Algunos resultados son:", lanzamientos[1:23])
```

```python
# Esta variable almacena nuestras la media conforme se suban el # de lanzamientos
media_acum = np.cumsum(lanzamientos) / np.arange(1,101)

# Creamos una dataframe de 2 columnas:
      # 1- Numero de lanzamientos
      # 2- Media acumulada
df = pd.DataFrame( {
    'Lanzamiento': np.arange(1,101),
    'Media Acumulada': media_acum } )

df.head(7)
```

Pasamos a graficar nuestras lanzamientos acumulados...

```python
plt.figure(figsize=(10,6))
plt.plot(df['Lanzamiento'], df['Media Acumulada'], label='media muestral')
plt.axhline(3.5, color= 'red', linestyle='--', label='media teórica (3.5)')
plt.xlabel('Número de lanzamientos')
plt.ylabel('Media')
plt.title('LGN en lanzamientos de un dado')
plt.legend()
plt.show()
```

## Problema de la Aguja de Bufon

Otra ejemplo de la utilidad que nos ofrece el *LGN* sera con el experimento a continuacion; donde este revela la existencia de $\pi$ anos antes de que se encontrara en las operaciones trigonometricas...

Para mas informacion: [**El problema que tiene el bufon con su agujas...**](https://www.estadisticaparatodos.es/taller/buffon/buffon.html)

```python
import numpy as np        # importamos esta paqueterias
import matplotlib.pyplot as plt
```

```python
# Parámetros del problema
L = 1.0  # Longitud de la aguja
d = 1.0  # Distancia entre las líneas paralelas
num_trials = 100000  # Número total de lanzamientos
step = 1000  # Número de lanzamientos por paso para la estimación de pi

# Listas para almacenar los valores de pi estimados y los lanzamientos
pi_estimates = []
trial_steps = []
```

```python
# Contador para cuántas agujas cruzan una línea
crossed_lines = 0

# Simulación de lanzamientos de la aguja y estimación de pi en pasos
for trial in range(1, num_trials + 1):
    # Posición del centro de la aguja
    center_position = np.random.uniform(0, d / 2)

    # Ángulo aleatorio entre 0 y pi (ángulo en radianes)
    angle = np.random.uniform(0, np.pi)

    # Distancia desde el centro de la aguja hasta la línea más cercana
    distance_to_nearest_line = L / 2 * np.sin(angle)

    # Verificamos si la aguja cruza una línea
    if center_position <= distance_to_nearest_line:
        crossed_lines += 1

    # Estimación de pi en pasos
    if trial % step == 0 and crossed_lines > 0:
        pi_estimate = (2 * L * trial) / (crossed_lines * d)
        pi_estimates.append(pi_estimate)
        trial_steps.append(trial)

print(f"Estimación de pi: {pi_estimate}")
```

```python
# Plot de la estimación de pi a lo largo de los lanzamientos
plt.figure(figsize=(10, 6))
plt.plot(trial_steps, pi_estimates, label="Estimación de π", color="blue")
plt.axhline(y=np.pi, color="red", linestyle="--", label="Valor real de π")
plt.xlabel('Número de lanzamientos')
plt.ylabel('Estimación de π')
plt.title('Convergencia de la estimación de π con el problema de la aguja de Buffon')
plt.legend()
plt.grid(True)
plt.show()
```

# Aplicación del TCL

Recordemos de forma muy resumida que **El Teorema del Limite Central:**

Sea un conjunto de **V.A.s** o **m.a.** *(DONDE DESCONOCEMOS LA DISTRIBUCION QUE MANEJA)*, pero existen $\mu_i$ y $\sigma^2_i$ *(Y SABEMOS QUE VALOR TIENE)* entonces:

$$ \sum_{i=0}^n X_i \sim N(0,1)$$

Veamos un ejemplo que nos visualice este hecho:

Sea $X_1,...,X_{10}$ una **m.a.** que se distribuya $Bin(1000,0.6)$, veamos que tiene un comportamiento cercano a una $N(0,1)$...

```python
# Crear el DataFrame con 50 muestras, cada una con 10 observaciones de una binomial(n=1000, p=0.6)
df = pd.DataFrame()

for i in range(1,51):
  muestra = np.random.binomial(1000,0.6,10)  # Muestra aleatoria de binomiales de 10 éxitos, se realiza 1000 veces el experimento
  col = f"muestra {i}"
  df[col] = muestra

df.head(5)  # Mostrar las primeras 5 filas
```

```python
# Calcular el promedio de cada muestra
df_muestra_medias = pd.DataFrame(df.mean(), columns=["Media de la muestra"])

# Visualizar la distribución de las medias
sns.distplot(df_muestra_medias)
#sns.histplot(df_muestra_medias["Media de la muestra"], kde=True)
plt.title("Distribución de medias muestrales")
plt.xlabel("Media")
plt.ylabel("Frecuencia")
plt.show()
```
