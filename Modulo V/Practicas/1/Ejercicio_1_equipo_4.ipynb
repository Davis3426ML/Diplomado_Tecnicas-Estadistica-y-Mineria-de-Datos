# Algoritmo

1: Encontrar la función de distribución acumulada:
$$
F(x) = \int_0^x 2t \, dt = \left[ t^2 \right]_0^x = x^2
$$
2: Entonces, la función de distribución es:
$$
F(x) = x^2 \quad \text{para } x \in (0,1)
$$
3: Invertir la función de distribución acumulada:

Sea $U \sim \text{Uniform}(0,1)$, queremos encontrar $X$ tal que $F(X) = U$:
$$
U = x^2 \quad \Rightarrow \quad x = \sqrt{U}
$$

### 1. Esperanza $ \mathbb{E}[X] $

Se define como:

$$
\mathbb{E}[X] = \int_0^1 x \cdot f(x) \, dx = \int_0^1 x \cdot 2x \, dx = \int_0^1 2x^2 \, dx
$$

Calculamos la integral:

$$
\mathbb{E}[X] = 2 \int_0^1 x^2 \, dx = 2 \left[ \frac{x^3}{3} \right]_0^1 = 2 \cdot \frac{1}{3} = \frac{2}{3}
$$

Resultado:

$$
\mathbb{E}[X] = \frac{2}{3}
$$

---

### 2. Varianza $ \mathrm{Var}(X) $

Primero necesitamos $ \mathbb{E}[X^2] $:

$$
\mathbb{E}[X^2] = \int_0^1 x^2 \cdot f(x) \, dx = \int_0^1 x^2 \cdot 2x \, dx = \int_0^1 2x^3 \, dx
$$

$$
\mathbb{E}[X^2] = 2 \int_0^1 x^3 \, dx = 2 \left[ \frac{x^4}{4} \right]_0^1 = 2 \cdot \frac{1}{4} = \frac{1}{2}
$$

Ahora aplicamos la fórmula de la varianza:

$$
\mathrm{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2 = \frac{1}{2} - \left( \frac{2}{3} \right)^2 = \frac{1}{2} - \frac{4}{9}
$$

$$
\mathrm{Var}(X) = \frac{9}{18} - \frac{8}{18} = \frac{1}{18}
$$

```python
import numpy as np

def generar_variable(n=1):
    u = np.random.uniform(0, 1, n)  # Paso 1: Generar U
    x = np.sqrt(u)                  # Paso 2: Aplicar la inversa F⁻¹(U)
    return x

# Ejemplo: Generar 10 valores
valores = generar_variable(10)
print(valores)
```
