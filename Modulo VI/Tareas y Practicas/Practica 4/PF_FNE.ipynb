```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar datos
file_path = "financial_news_events.csv"
df = pd.read_csv(file_path)

# Vista general
print("Dimensiones:", df.shape)
print("\nTipos de datos:")
print(df.dtypes)

print("\nValores faltantes por columna:")
print(df.isnull().sum())

print("\nDescripción estadística (numéricas):")
print(df.describe())

print("\nDescripción estadística (categóricas):")
print(df.describe(include="object"))

# =========================
# DISTRIBUCIONES
# =========================

# Distribución de cambios en índice
plt.figure(figsize=(8,5))
sns.histplot(df["Index_Change_Percent"].dropna(), bins=30, kde=True)
plt.title("Distribución de cambios porcentuales en índices")
plt.xlabel("Cambio %")
plt.ylabel("Frecuencia")
plt.show()

# Volumen de trading
plt.figure(figsize=(8,5))
sns.boxplot(x=df["Trading_Volume"])
plt.title("Boxplot: Volumen de trading")
plt.show()

# Sentimiento
plt.figure(figsize=(6,4))
df["Sentiment"].value_counts().plot(kind="bar", color="skyblue")
plt.title("Frecuencia de Sentimientos en Noticias")
plt.xlabel("Sentimiento")
plt.ylabel("Número de noticias")
plt.show()

# Eventos de mercado más frecuentes
plt.figure(figsize=(10,5))
df["Market_Event"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Eventos de Mercado más reportados")
plt.show()

# Sectores más frecuentes
plt.figure(figsize=(10,5))
df["Sector"].value_counts().head(10).plot(kind="bar", color="orange")
plt.title("Top 10 Sectores más frecuentes en noticias")
plt.show()

# Impacto
plt.figure(figsize=(6,4))
df["Impact_Level"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Nivel de impacto reportado")
plt.show()
```
