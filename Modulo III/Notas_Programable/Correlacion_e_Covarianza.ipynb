El campo de control de contaminates en cuerpos acuáticos se realiza principalmente por la industria textil y peletera. Sus desechos son químicamente complejos: caracterizados por valores elevados de demanda de oxigeno químico, solidos volátiles y otros containantes.

Se obtuvieron 33 unidades experimentales de desechos tratados químicamente en un estudio realizado sobre el Río Lerma. Se registro la reducción porcentual de solidos totales y el procentaje de disminución de la demanda de oxigeno quimico *(Archivo: ContamH2Opeletera393.xlsx ¡Abrirlo!)*


```python
#===================================
# Importamos los modulos a utilizar
#===================================
import numpy as np
import pandas as pd
```

```python
# Manipulación de bases de datos '.csv' .'xlsx' con pandas

#==========================
# Cargamos la base de datos (Los valores cualitativos ya estan tratados en la base como cuantitativo)
#==========================

base3 = pd.read_excel("ContamH2Opeletera393.xlsx")
base3.head(12)
```

Hay un par de variables, ¿estarán relacionadas?. Ello lo podemos determinar de dos maneras:

- Númericamente: de las cuales podemos usar 2 tipos de correlacion...

```python
# Correlacion de spearman
corSp = base3.corr( method='spearman')
print(corSp)
```

```python
# Correlacion de Pearson
matriz = base3.corr(method='pearson')
print(matriz)
```

Ambas correlacion son buenas en lo que hace pero podemos notar que hay una diferencia que hay entre ambos resultados. Esto se debe al hecho ha que se dedica cada una de ellas...

**Coeficiente de Correlacion de Spearman**

Esta correlacion de encarga de ver la relacion de los datos entre ellos y no tanto en el hecho de que provenga de un par de *V.A.*

- Acepta valores atipocos
- no es necesario que los valores se comporte como una normal
- que los valores tenga sentido alguno con el estudio

**Coeficiente de Correlacion de Pearson**

Este tiene el objetivo de establecer la relacion de forma mas estandar de las **V.A.** de manera lineal

- Es todo lo contrario a la de spearman

# Gráfica de dispersión

```python
# Paqueteria para realizar graficas
import matplotlib.pyplot as plt

#El tamano del lienzo: 12 largo, 8 altura
plt.figure(figsize = (10,7))

# Una grafica de Puntos
plt.scatter( x = base3['Reduc_solidos'], y = base3['Red_deman_oxig%'], alpha = 0.8, label = 'XY')

#etiquetado
plt.xlabel('Reduccion de Solidos'); plt.ylabel('La demanda de Oxigeno');
plt.title('Relacion de los Datos')
plt.legend()
plt.grid(True)
# incluir la tendencia tentativa
plt.show()
```
